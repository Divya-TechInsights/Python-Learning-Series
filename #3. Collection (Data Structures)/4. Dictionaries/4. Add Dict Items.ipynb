## Add Dictionary Items

### Adding a new key-value pair
Adding an item to the dictionary is done by using a new index key and assigning a value to it

```python
person = {"name": "Alice", "age": 25}

# Add a new key
person["city"] = "London"
print(person)
```

If new_key does not exist, it will be added.

If new_key already exists, its value will be updated.

### Updating an existing key

```python
person = {"name": "Alice", "age": 25}

person["age"] = 26  # age already exists
print(person)
```

### Update Dictionary
The update() method will update the dictionary with the items from a given argument. If the item does not exist, the item will be added.

The argument must be a dictionary, or an iterable object with key:value pairs.



The update() method allows you to modify a dictionary by adding or updating items from another source.

#### Key points:

Update existing keys → their values are overwritten.

Add new keys → if the key doesn’t exist, it’s added automatically.

#### Argument types:

1. Another dictionary

2. An iterable of key-value pairs (like list of tuples, set of tuples, generator)

3. Keyword arguments (optional shortcut)

### 1. Another Dictionary

```python
_dict = {"brand": "Ford", "model": "Mustang", "year": 1964}

_dict.update({"color": "red"})
print(_dict)
```

### Adding multiple items using update

```python
person = {"name": "Alice", "age": 25}

person.update({"country": "UK", "profession": "Engineer"})
print(person)
```

Key point:

dict[key] = value is great for adding single items.

update() is better for adding multiple items at once.

### 2. An iterable of key-value pairs (like list of tuples, set of tuples, generator)

#### Updating with list of tuples

```python
person = {"name": "Alice", "age": 25}

person.update([("country", "UK"), ("profession", "Engineer")])
print(person)
```

#### Updating with keyword arguments

```python
person = {"name": "Alice", "age": 25}

person.update(state="England", hobby="Reading")
print(person)
```

update() modifies the original dictionary, it does not create a new one.

Use dict[key] = value for single items, update() for multiple items at once.
