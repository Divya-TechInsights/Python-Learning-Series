### Python Dictionaries


| Feature             | Dictionary Description                                      |
| ------------------- | ----------------------------------------------------------- |
| Purpose             | Stores data values in **key : value** pairs                 |
| Type                | Collection data type                                        |
| Order               | **Ordered** (from Python 3.7 onward)                        |
| Mutability          | **Changeable (mutable)**                                    |
| Duplicate Keys      | **Not allowed**                                             |
| Duplicate Values    | Allowed                                                     |
| Syntax              | Written using **curly brackets `{}`**                       |
| Structure           | `{key1: value1, key2: value2}`                              |
| Python Version Note | Python 3.6 and earlier → unordered<br>Python 3.7+ → ordered |



### Create a dictionary

```python
_dict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(_dict)
```

### Dictionary Items

Dictionary items are ordered, changeable, and do not allow duplicate keys.

Each item in a dictionary is stored as a key : value pair.

You can access the value by using its key name.

```python
_dict = {"brand": "Ford", "model": "Mustang", "year": 1964}

print(_dict["brand"])
```

| Concept                  | Explanation                                                                  |
| ------------------------ | ---------------------------------------------------------------------------- |
| Ordered (Python ≥ 3.7)   | Dictionary items have a **defined order**, and the order **does not change** |
| Unordered (Python ≤ 3.6) | Dictionary items **do not have a defined order**                             |
| Index Access             | Dictionaries **cannot be accessed using index numbers**                      |
| Changeable               | **Values can be changed**, items can be **added or removed**                 |
| Key Change               | **Keys cannot be changed** (they are immutable)                              |
| Duplicate Keys           | **Not allowed** — each key must be unique                                    |
| Duplicate Values         | Allowed                                                                      |


### Dictionary is mutable.

This means:

You can change values

You can add new key–value pairs

You can remove existing items

Keys must be immutable (int, string, tuple), but values can be anything

Note - Dictionary is a mutable data type in Python.

### Duplicate Values Will Overwrite Existing Values

In a dictionary, duplicate keys are not allowed.

If the same key appears more than once, the last value assigned to that key overwrites the previous one.

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964, "year": 2026}

print(thisdict)
```

### Key point:

Duplicate keys →  not allowed

Last value of the duplicate key → overwrites earlier value
