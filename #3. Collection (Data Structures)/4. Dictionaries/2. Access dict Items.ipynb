### Accessing Dictionary Items

You can access dictionary items by referring to the key name inside square brackets [].

```python
_dict = {"brand": "Ford", "model": "Mustang", "year": 1964}

x = _dict["model"]
print(x)
```

note - If the key does not exist, this method will raise an error.

### Using the get() Method

The get() method also returns the value of a key, but it is safer because it does not raise an error if the key is missing.

```python
_dict = {"brand": "Ford", "model": "Mustang", "year": 1964}
x = _dict.get("model")
print(x)
```
