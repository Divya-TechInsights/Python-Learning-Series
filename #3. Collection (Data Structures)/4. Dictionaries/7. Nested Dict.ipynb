## Nested Dictionaries
A nested dictionary means a dictionary inside another dictionary.

#### 1. Direct Creation
Create a dictionary that contain other dictionaries

```python
# Create a nested dictionary
company = {"employee1": {"name": "Amit", "age": 28, "department": "IT"},
"employee2": {"name": "Neha", "age": 25, "department": "HR"}}

# Print the nested dictionary
print(company)
```

```python
family = {"child1" : {"name" : "Emma", "year" : 2004},
  "child2" : {"name" : "Thomas", "year" : 2007},
  "child3" : {"name" : "Lucy","year" : 2011}}

print(family)
```

### 2. Create First, Then Nest 
Create three dictionaries, then create one dictionary that will contain the other three dictionaries

```python
# Create three dictionaries
student1 = {"name": "Rahul","age": 20,"marks": 85}

student2 = {"name": "Anita", "age": 21, "marks": 90}

student3 = {"name": "Vikas", "age": 19, "marks": 78}

# Create one dictionary containing the above three dictionaries
students = {"student1": student1, "student2": student2, "student3": student3}

# Print the nested dictionary
print(students)
```

```python
child1 = {"name" : "Emma", "year" : 2004}
child2 = {"name" : "Thomas", "year" : 2007}
child3 = {"name" : "Lucy", "year" : 2011}

myfamily = {"child1" : child1, "child2" : child2, "child3" : child3}
print(myfamily)
```

### 3. Access Nested Dictionary Values

```python
student = {"name": "Rahul", "marks": {"math": 80, "science": 75}}

print(student["name"])
print(student["marks"]["math"])
print(student["marks"]["science"])
```

### 4. Modify Nested Dictionary Values

```python
student = {"name": "Rahul", "marks": {"math": 80, "science": 75}}
student["marks"]["math"] = 90
student["marks"]["english"] = 85

print(student)
```

### 5. Delete Nested Dictionary Values

```python
student = {"name": "Rahul", "marks": {"math": 80, "science": 75}}
del student["marks"]["science"]

print(student)
```

### 6. Shallow Copy of Nested Dictionary

```python
student1 = {"name": "Rahul", "marks": {"math": 80, "science": 75}}
student2 = student1.copy()
student2["marks"]["math"] = 100

print(student2)
```
