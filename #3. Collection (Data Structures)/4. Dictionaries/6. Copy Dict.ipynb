## Copy a Dictionary

### Using copy() method

```python
dict1 = {"a": 1, "b": 2, "c": 3}

dict2 = dict1.copy()

print(dict2)
```

```python
dict1 = {"brand": "Ford", "model": "Mustang", "year": 1964}
dict2 = dict1.copy()
print(dict2)
```

### Using dict() constructor

```python
dict1 = {"a": 1, "b": 2}

dict2 = dict(dict1)
print(dict2)
```

```python
dict1 = {"brand": "Ford", "model": "Mustang", "year": 1964}

dict2 = dict(dict1)
print(dict2)
```
