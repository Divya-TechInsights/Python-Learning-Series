## Dictionary Methods

| Method         | Description                                                                           |
| -------------- | ------------------------------------------------------------------------------------- |
| `clear()`      | Removes **all elements** from the dictionary                                          |
| `copy()`       | Returns a **shallow copy** of the dictionary                                          |
| `fromkeys()`   | Creates a new dictionary using specified **keys** and a **single value**              |
| `get()`        | Returns the value of the specified key (**no error if key not found**)                |
| `items()`      | Returns a view object containing **(key, value)** pairs                               |
| `keys()`       | Returns a view object containing **all keys**                                         |
| `pop()`        | Removes the element with the **specified key**                                        |
| `popitem()`    | Removes and returns the **last inserted key–value pair**                              |
| `setdefault()` | Returns the value of a key; if key doesn’t exist, **inserts it with a default value** |
| `update()`     | Updates the dictionary with **key–value pairs** from another dictionary               |
| `values()`     | Returns a view object containing **all values**                                       |


### 1. clear()

Removes all elements from the dictionary (dictionary stays, data gone)

```python
d = {"a": 1, "b": 2}
d.clear()
print(d)   # {}
```

### 3. copy()

Returns a shallow copy of the dictionary

```python
d = {"a": 1, "b": 2}
new_d = d.copy()
print(new_d)
```

### 3. fromkeys()

Creates a new dictionary using given keys and same value for all

```python
keys = ("a", "b", "c")
d = dict.fromkeys(keys, 100)
print(d)   # {'a': 0, 'b': 0, 'c': 0}
```

### 4. get()

Returns value of a key

No error if key is missing

```python
d = {"a": 1}
print(d.get("a"))      # 1
print(d.get("b"))      # None
```

### 5. items()

Returns key–value pairs as tuples

```python
d = {"a": 1, "b": 2}
print(d.items())
```

### 6. keys()

Returns all keys

```python
d = {"a": 1, "b": 2}
print(d.keys())
```

### 7. pop()

Removes and returns value of specified key

```python
d = {"a": 1, "b": 2}
value = d.pop("a")
print(d)       # {'b': 2}
```

### 8. popitem()

Removes and returns the last inserted key-value pair

```python
d = {"a": 1, "b": 2}
item = d.popitem()
print(d)
```

### 9. setdefault()

Returns value if key exists

If not → adds key with default value

```python
d = {"a": 1}
d.setdefault("b", 10)
print(d)   # {'a': 1, 'b': 10}
```

### 10. update()

Updates dictionary using another dictionary or key-value pairs

```python
d = {"a": 1}
d.update({"b": 2, "c": 3})
print(d)
```

### 11. values()

Returns all values

```python
d = {"a": 1, "b": 2}
print(d.values())
```
