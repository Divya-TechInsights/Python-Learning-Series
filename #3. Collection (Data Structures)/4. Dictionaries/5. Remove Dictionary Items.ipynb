## Remove Dictionary Items

#### The pop() method 
It removes the item with the specified key name

```python
person = {"name": "Alice", "age": 25, "city": "London"}

person.pop("city")
print(person)
```

#### The popitem() method 
It removes the last inserted item (in versions before 3.7, a random item is removed instead)

Removes the last inserted key-value pair (Python 3.7+ dictionaries maintain insertion order).

```python
person = {"name": "Alice", "age": 25, "city": "London"}

person.popitem()
print(person)
```

#### Using del

Removes a specific key permanently.

```python
person = {"name": "Alice", "age": 25, "city": "London"}

del person["age"]
print(person)
```

Raises KeyError if the key doesn’t exist.

Can also delete the entire dictionary: del person

### Using clear()

Removes all items from the dictionary.

```python
person = {"name": "Alice", "age": 25, "city": "London"}

person.clear()
print(person)
```

```python
d = {}
d.clear()   # No error
print(d)
```

Error:  No (even if the dictionary is already empty)

clear() empties the dictionary but keeps the dictionary object

| Method      | Removes            | Returns Value? | Error if key missing?      |
| ----------- | ------------------ | -------------- | -------------------------- |
| `pop(key)`  | Specific key       | Yes            | Yes (unless default given) |
| `popitem()` | Last inserted item | Yes            | No (empty dict error)      |
| `del key`   | Specific key       | No             | Yes                        |
| `clear()`   | All items          | No             | No                         |
