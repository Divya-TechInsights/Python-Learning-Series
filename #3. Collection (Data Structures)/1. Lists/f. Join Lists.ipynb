### Join Two Lists

There are several ways to join, or concatenate, two or more lists in Python

### 1. Using + Operator

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

list3 = list1 + list2
print(list3)
```

### 2. Using extend() method

```python
list1 = ["a", "b" , "c"]
list2 = [1, 2, 3]

list1.extend(list2)
print(list1)
```
