### Add List Items

### 1. Append Items

To add an item to the end of the list

```python
list = ["apple", "banana", "cherry"]
list.append("orange")
print(list)
```

```python
list = [1, 2, 3]
list.append(4)
print(list)
```

### 2. Insert Items

To insert a list item at a specified index

```python
list = ["apple", "banana", "cherry"]
list.insert(1, "orange")
print(list)
```

```python
list = [4, 5, 6, 7]
list.insert(0, 8)
print(list)
```

### 3. Extend Items

Add multiple items from another list

```python
list = ["apple", "banana", "cherry"]
list1 = ["mango", "pineapple", "papaya"]
list.extend(list1)
print(list)
```

Note: The extend() method can add elements from any iterable object, not only lists.

#### What is an iterable? 

An iterable is anything that contains multiple items, like:

list

tuple

set

dictionary

string

Python can go through these one by one.

#### What extend() actually does?

It takes items one by one from any iterable and adds them to the list.

#### Using a Tuple

```python
mylist = [1, 2]
mylist.extend((3, 4))
print(mylist)
```

#### Using a set

```python
mylist = [1, 2]
mylist.extend({3, 4})
print(mylist)
```

#### Using a dictionary

```python
mylist = [1, 2]
mylist.extend({"a": 10, "b": 20})
print(mylist)
```

Only dictionary keys are added (by default).

#### Using a string

```python
mylist = []
mylist.extend("hello")
print(mylist)
```

### Compare extend() with append()

| Feature               | `append()`     | `extend()`            |
| --------------------- | -------------- | --------------------- |
| What it adds          | **One item**   | **Multiple items**    |
| Adds list as          | Single element | Elements of the list  |
| Accepts               | Any object     | Only iterable objects |
| Changes original list | Yes            | Yes                   |
| Common use            | Add one value  | Add many values       |


### Example

Using append()

```python
mylist = [1, 2, 3]
mylist.append([4, 5])
print(mylist)
```

```python
mylist = [1, 2, 3]
tuple = (4, 5)
mylist.append(tuple)
print(mylist)
```

### Example

Using extend()

```python
mylist = [1, 2, 3]
mylist.extend([4, 5])
print(mylist)
```

```python
list = [1, 2, 3]
tuple = (4, 5)
list.extend(tuple)
print(list)
```
