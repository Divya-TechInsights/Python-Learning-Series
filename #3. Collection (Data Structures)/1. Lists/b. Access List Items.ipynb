### Access Items

List items are indexed and can be accessed by referring to the index number

### 1️. Indexing in Lists

- Every item in a list has a position number, called an index.

- Python indexing starts at 0, meaning the first item has index 0, the second item has index 1, and so on.

<br>

fruits = ["apple", "banana", "cherry", "mango"]

| Item   | Index |
| ------ | ----- |
| apple  | 0     |
| banana | 1     |
| cherry | 2     |
| mango  | 3     |


2️. Accessing Items by Index

list_name[index]

```python
fruits = ["apple", "banana", "cherry", "mango"]
print(fruits[0])  # apple → first item
print(fruits[2])  # cherry → third item
```

3️. Negative Indexing

Python also allows negative indexes, which start counting from the end.

-1 → last item, -2 → second last, etc.

```python
fruits = ["apple", "banana", "cherry", "mango"]
print(fruits[-1])  # mango → last item
print(fruits[-2])  # cherry → second last item
```

4️. Accessing a Range of Items (Slicing)

We can access multiple items using slicing.

fruits[1:3]    

items from index 1 to 2 (3 not included)

```python
fruits = ["apple", "banana", "cherry", "apple", "cherry"]
print(fruits[1:3])  # ['banana', 'cherry']
print(fruits[:2])   # ['apple', 'banana'] → from start to index 1
print(fruits[2:])   # ['cherry', 'mango'] → from index 2 to end
print(fruits[-3:-1])  # ['cherry', 'apple'] → from index -3 to -1
print(fruits[-4:])    # ['banana', 'cherry', 'apple', 'cherry'] → from index -4 to end
```

### Key Points

- Indexing starts at 0.

- Negative index counts from the end.

- Slicing allows accessing a range of items.
