### Copy Lists

1. Use the built-in List method copy() to copy a list

```python
list = ["apple", "banana", "cherry"]
mylist = list.copy()
print(mylist)
```

### 2. make a copy of a list by using the : (slice) operator

```python
thislist = ["apple", "banana", "cherry"]
mylist = thislist[:]
print(mylist)
```
