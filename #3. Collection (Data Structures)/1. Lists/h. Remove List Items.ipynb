### Remove List Items

#### 1. remove() – Remove by value

```python
list = ["apple", "banana", "cherry"]
list.remove("banana")
print(list)
```

#### If there are more than one item with the specified value, the remove() method removes the first occurrence

```python
list = ["banana","apple", "banana", "cherry", "banana", "kiwi"]
list.remove("banana")
print(list)
```

### 2. pop() – Remove by index

```python
fruits = ["apple", "banana", "cherry"]
fruits.pop(1)
print(fruits)
```

Remove last item - pop()

```python
fruits = ["apple", "banana", "cherry"]
fruits.pop()
print(fruits)
```

### 3. del keyword – Remove by index or slice

```python
fruits = ["apple", "banana", "cherry"]
del fruits[0]
print(fruits)
```

delete multiple items

```python
fruits = ["apple", "banana", "cherry"]
del fruits[:]  # delete the entire list
print(fruits)

fruits = ["apple", "banana", "cherry"]
del fruits[0:2]  # delete a range of items
print(fruits)
```

### 4. clear() – Remove all items

```python
fruits = ["apple", "banana"]
fruits.clear()
print(fruits)
```

| Method     | Removes by    | Removes how many |
| ---------- | ------------- | ---------------- |
| `remove()` | Value         | First match      |
| `pop()`    | Index         | One              |
| `del`      | Index / slice | One or many      |
| `clear()`  | —             | All              |
