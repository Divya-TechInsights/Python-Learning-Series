### Change List Items

### 1. Change the single item

```python
list = ["apple", "banana", "cherry"]
list[1] = "blackcurrant"
print(list)
```

### 2. Change a range of item values

```python
list = ["apple", "banana", "cherry", "orange", "kiwi", "mango"]
list[1:3] = ["blackcurrant", "watermelon"]
print(list)
```

### 3. Updating list elements with multiple values

```python
list = ["apple", "banana", "cherry"]
list[1:2] = ["blackcurrant", "watermelon"]
print(list)
```
