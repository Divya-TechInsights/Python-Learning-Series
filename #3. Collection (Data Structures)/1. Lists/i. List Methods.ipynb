### List Methods

| Method      | Description                                              |
| ----------- | -------------------------------------------------------- |
| `append()`  | Adds an element at the end of the list                   |
| `clear()`   | Removes all elements from the list                       |
| `copy()`    | Returns a copy of the list                               |
| `count()`   | Returns the number of elements with the specified value  |
| `extend()`  | Adds elements of another iterable to the end of the list |
| `index()`   | Returns the index of the first matching element          |
| `insert()`  | Adds an element at the specified position                |
| `pop()`     | Removes the element at the specified position            |
| `remove()`  | Removes the first occurrence of the specified value      |
| `reverse()` | Reverses the order of the list                           |
| `sort()`    | Sorts the list                                           |


### 1. append() method

#### Syntax
     list.append(element)


#### Parameter Values
| Parameter | Description                                                                                                              |
| --------- | ------------------------------------------------------------------------------------------------------------------------ |
| `element` | **Required.** The item you want to add to the list. It can be of **any data type** (string, number, list, object, etc.). |


```python
fruits = ['apple', 'banana', 'cherry']
fruits.append("orange")
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'orange']
```

### 2. clear() method

| **Category**         | **Description** |
| -------------------- | --------------- |
| **Syntax**           | `list.clear()`  |
| **Parameter Values** | No parameters   |
| **Return Value**     | Returns `None`  |


```python
fruits = ['apple', 'banana', 'cherry', 'orange']
fruits.clear()
print(fruits)  # Output: []
```

### 3. copy() method

| **Category**         | **Description**                        |
| -------------------- | -------------------------------------- |
| **Syntax**           | `list.copy()`                          |
| **Parameter Values** | No parameters                          |
| **Return Value**     | Returns a **copy** of the list         |


```python
fruits = ['apple', 'banana', 'cherry', 'orange']
x = fruits.copy()
print(x)  # Output: ['apple', 'banana', 'cherry', 'orange']
```

### 4. count() method

| **Category**              | **Description**                                                                              |
| ------------------------- | -------------------------------------------------------------------------------------------- |
| **Syntax**                | `list.count(value)`                                                                          |
| **Parameter**             | `value`                                                                                      |
| **Parameter Description** | Required. Any type (string, number, list, tuple, etc.). The value to search for in the list. |
| **Return Value**          | Returns the **number of times** the specified value appears in the list                      |


```python
fruits = ["apple", "banana", "apple", "orange", "apple"]
result = fruits.count("apple")
print(result)  # Output: 3
```

### 5. extend() method

| **Category**              | **Description**                                                                                  |
| ------------------------- | ------------------------------------------------------------------------------------------------ |
| **Syntax**                | `list.extend(iterable)`                                                                          |
| **Parameter**             | `iterable`                                                                                       |
| **Parameter Description** | Required. Any iterable (list, set, tuple, string, etc.) whose elements will be added to the list |
| **Return Value**          | Returns **None**                                                                                 |


```python
numbers = [1, 2, 3]
numbers.extend([4, 5, 6])
print(numbers) # Output: [1, 2, 3, 4, 5, 6]
```

### 6. index() method

| **Category**              | **Description**                                                            |
| ------------------------- | -------------------------------------------------------------------------- |
| **Syntax**                | `list.index(elmnt, start, end)`                                            |
| **Parameter**             | `elmnt`                                                                    |
| **Parameter Description** | Required. Any type (string, number, list, etc.). The element to search for |
| **Start**                 | Optional. Index number where the search begins                             |
| **End**                   | Optional. Index number where the search ends                               |
| **Return Value**          | Returns the **index position** of the first matching element               |


```python
numbers = [10, 20, 30, 20, 40]
result = numbers.index(20)
print(result)
# Output: 1 (the index of the first occurrence of 20)
```

```python
# Example with start and end parameters
numbers = [10, 20, 30, 20, 40]
result = numbers.index(20, 2, 4)
print(result)
# Output: 3 (the index of the first occurrence of 20 between index 2 and 5)
```

```python
numbers = [10, 20, 30, 20, 40]
result = numbers.index(20, 2)
print(result)
```

### 7. insert() method

| **Category**              | **Description**                                                               |
| ------------------------- | ----------------------------------------------------------------------------- |
| **Syntax**                | `list.insert(pos, elmnt)`                                                     |
| **Parameter**             | `pos`                                                                         |
| **Parameter Description** | Required. A number specifying the position where the element will be inserted |
| **Element**               | `elmnt`                                                                       |
| **Element Description**   | Required. An element of any type (string, number, object, etc.)               |
| **Return Value**          | Returns **None**                                                              |


```python
numbers = [10, 20, 30]
numbers.insert(1, 15)
print(numbers)  # Output: [10, 15, 20, 30]
```

### 8. pop() method

| **Category**              | **Description**                                                                                                           |
| ------------------------- | ------------------------------------------------------------------------------------------------------------------------- |
| **Syntax**                | `list.pop(pos)`                                                                                                           |
| **Parameter**             | `pos`                                                                                                                     |
| **Parameter Description** | Optional. A number specifying the position of the element to remove. Default is `-1` (removes and returns the last item). |
| **Return Value**          | Returns the **removed element**                                                                                           |


```python
numbers = [10, 20, 30, 40]
numbers1 = numbers.pop(2)
print(numbers1)
print(numbers) # Output: 30
```

```python
### Example (default parameter)
numbers = [10, 20, 30, 40]
removed = numbers.pop()
print(removed)
print(numbers) # Output: 40
```

### 9. remove() method

| **Category**              | **Description**                                                                               |
| ------------------------- | --------------------------------------------------------------------------------------------- |
| **Syntax**                | `list.remove(elmnt)`                                                                          |
| **Parameter**             | `elmnt`                                                                                       |
| **Parameter Description** | Required. Any type (string, number, list, etc.). The element you want to remove from the list |
| **Return Value**          | Returns **None**                                                                              |


```python
fruits = ["apple", "banana", "orange", "banana"]
fruits.remove("banana")
print(fruits)  # Output: ['apple', 'orange', 'banana']
```

### Comparison remove() vs pop()


| **Feature**            | **`remove()`**                        | **`pop()`**                              |
| ---------------------- | ------------------------------------- | ---------------------------------------- |
| **Purpose**            | Removes an element **by value**       | Removes an element **by index**          |
| **Syntax**             | `list.remove(elmnt)`                  | `list.pop(pos)`                          |
| **Parameter**          | Element value (required)              | Index position (optional)                |
| **Default Behavior**   | No default                            | Default index = `-1` (last element)      |
| **Return Value**       | Returns `None`                        | Returns the **removed element**          |
| **Removes**            | First matching occurrence             | Element at given index                   |
| **Error if not found** | `ValueError`                          | `IndexError`                             |
| **Common Use Case**    | When you know the **value** to delete | When you know the **position** to delete |


### 9. reverse() method

| **Category**         | **Description**  |
| -------------------- | ---------------- |
| **Syntax**           | `list.reverse()` |
| **Parameter Values** | No parameters    |
| **Return Value**     | Returns **None** |


```python
numbers = [1, 2, 3, 4]
numbers.reverse()
print(numbers)  # Output: [4, 3, 2, 1]
```

### 10. sort() method

| **Category**              | **Description**                                                                              |
| ------------------------- | -------------------------------------------------------------------------------------------- |
| **Syntax**                | `list.sort(reverse=True)`                                                                    |
| **Parameter**             | `reverse`                                                                                    |
| **Parameter Description** | Optional. `reverse=True` sorts the list in **descending order**. Default is ascending order. |
| **Return Value**          | Returns **None** (sorts the list in place)                                                   |

        
      


```python
numbers = [4, 1, 3, 2]

# Ascending (default)
numbers.sort()
print(numbers)

# Descending
numbers.sort(reverse=True)
print(numbers)
```
