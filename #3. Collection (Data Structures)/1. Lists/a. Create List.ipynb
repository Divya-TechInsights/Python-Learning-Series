### Lists

- Lists are used to store multiple items in a single variable.

- List items are ordered, changeable, and allow duplicate values.


Explanation:

fruits = ["apple", "banana", "cherry"]





| Property          | Meaning                              | Example                 |
| ----------------- | ------------------------------------ | ----------------------- |
| Ordered           | Items have a fixed position          | `fruits[0]` → `"apple"` |
| Changeable        | Can modify, add, remove items        | `fruits[1] = "orange"`  |
| Allows Duplicates | Same value can appear multiple times | `[1, 2, 2, 3]`          |


### Create a list

```python
list = ["apple", "banana", "cherry"]
print(list)
```

### List allow duplicate values

```python
list = ["apple", "banana", "cherry", "apple", "cherry"]
print(list)
```

### Length of List

```python
list = ["apple", "banana", "cherry", "cherry"]
print(len(list))
```

## Use of List in Real Life 

A **list** is used when you need to store **many items together** and **change them easily**.

Think of a **list as a digital notebook** that you can edit anytime.



## Real-Life Uses of List

### 1️ Student Marks List

A teacher stores marks of students.

* Marks can be updated
* New student can be added
* Wrong entry can be removed

Because data **keeps changing**, we use a list.



### 2️ Shopping Cart

When you shop online:

* Add items
* Remove items
* Change quantity

Your cart is a **list** because it is editable.



### 3️ To-Do List App

* Add new tasks
* Delete completed tasks
* Reorder tasks

A perfect example of a list.



### 4️ Playlist in Music App

* Add songs
* Remove songs
* Shuffle order

A playlist is a list.



### 5️ Attendance Register

Students join, leave, get corrected

So attendance is stored in a list.



### 6️ Machine Learning Dataset

Feature values, labels, records

Stored in lists before converting to arrays



## Summary

> Use a **list** when your data needs to **change, grow, or shrink**.
