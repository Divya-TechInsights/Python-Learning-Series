## Sets

| Feature                    | Description                                                |
| -------------------------- | ---------------------------------------------------------- |
| **What is a Set?**         | A set is used to store multiple items in a single variable |
| **Data Type Category**     | One of the 4 built-in collection data types in Python      |
| **Other Collection Types** | List, Tuple, Dictionary                                    |
| **Order**                  | Unordered (items do not have a fixed position)             |
| **Indexing**               | Unindexed (items cannot be accessed using index numbers)   |
| **Changeability**          | Items are unchangeable (immutable values)                  |
| **Add Items**              | Yes, new items can be added                                |
| **Remove Items**           | Yes, items can be removed                                  |
| **Duplicate Values**       | Not allowed (each item must be unique)                     |
| **Syntax Example**         | `{1, 2, 3, 4}`                                             |


Note - A set is a collection which is unordered, unchangeable*, and unindexed.

* Note: Set items are unchangeable, but you can remove items and add new items.

### Create a Set

```python
_set = {"apple", "banana", "cherry"}
print(_set)
```

Note: Sets are unordered, so you cannot be sure in which order the items will appear.
