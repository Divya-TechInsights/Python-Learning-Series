### Set Methods


| Method                          | Shortcut | Description                                             |                                
| ------------------------------- | -------- | ------------------------------------------------------- | 
| `add()`                         | —        | Adds an element to the set                              |                                
| `clear()`                       | —        | Removes all elements from the set                       |                                
| `copy()`                        | —        | Returns a shallow copy of the set                       |                                
| `difference()`                  | `-`      | Returns elements present in first set but not in others |                                
| `difference_update()`           | `-=`     | Removes common elements from the original set           |                                
| `discard()`                     | —        | Removes specified element (no error if not found)       |                                
| `intersection()`                | `&`      | Returns common elements of sets                         |                                
| `intersection_update()`         | `&=`     | Keeps only common elements in the original set          |                                
| `isdisjoint()`                  | —        | Returns `True` if no common elements exist              |                                
| `issubset()`                    | `<=`     | Returns `True` if set is subset of another              |                                
| `issubset()`                    | `<`      | Returns `True` if set is proper subset                  |                                
| `issuperset()`                  | `>=`     | Returns `True` if set is superset of another            |                                
| `issuperset()`                  | `>`      | Returns `True` if set is proper superset                |                                
| `pop()`                         | —        | Removes and returns a random element                    |                                
| `remove()`                      | —        | Removes specified element (error if not found)          |                                
| `symmetric_difference()`        | `^`      | Returns elements not common to both sets                |                                
| `symmetric_difference_update()` | `^=`     | Updates set with symmetric difference                   |                                
| `union()`                       |       | Returns union of sets                                                         | 
| `update()`                      |       |                                                       Updates set with union of sets |


### 1. add()

```python
s = {1, 2}
s.add(3)
print(s)
```

### 2. clear()

```python
s = {1, 2, 3}
s.clear()
print(s)
```

### 3. copy()

```python
s = {10, 20}
c = s.copy()
print(c)
```

### 4. difference() (-)

```python
a = {1, 2, 3}
b = {3, 4}

print(a.difference(b))
print(a - b)
```

### 5. difference_update() (-=)

```python
a = {1, 2, 3, 4}
b = {3, 4}

a.difference_update(b)
print(a)
```

| Feature                      | `difference()` | `difference_update()` |
| ---------------------------- | -------------- | --------------------- |
| Returns a new set            | Yes            | No                    |
| Modifies original set        | No             | Yes                   |
| Original set after operation | Unchanged      | Changed               |
| Can use operator             | `-`            | `-=`                  |


```python
a = {1, 2, 3, 4}
b = {3, 4}

a -= b
print(a)
```

### 6. discard()

```python
s = {1, 2, 3}
s.discard(2)
print(s)
```

### 7. intersection() (&)

```python
a = {1, 2, 3}
b = {2, 3, 4}

print(a.intersection(b))
print(a & b)
```

### 8. intersection_update() (&=)

```python
a = {1, 2, 3, 4}
b = {3, 4}

a.intersection_update(b)
print(a)
```

```python
a = {1, 2, 3, 4}
b = {3, 4}

a &= b
print(a)
```

| Feature                      | `intersection()` | `intersection_update()` |
| ---------------------------- | ---------------- | ----------------------- |
| Returns a new set            | Yes              | No                      |
| Modifies original set        | No               | Yes                     |
| Original set after operation | Unchanged        | Changed                 |
| Can use operator             | `&`              | `&=`                    |


### 9. isdisjoint()

```python
x = {1, 2}
y = {3, 4}

print(x.isdisjoint(y))
```

### 10. issubset() (<=, <)

```python
a = {1, 2}
b = {1, 2, 3}

print(a.issubset(b))  # True
print(a <= b)         # True
print(a < b)          # True
```

### 11. issuperset() (>=, >)

```python
a = {1, 2, 3}
b = {1, 2}

print(a.issuperset(b))  # True
print(a >= b)           # True
print(a > b)            # True
```

### 12. pop()

```python
s = {10, 20, 30}
s.pop()
print(s)
```

### 13. remove()

```python
s = {1, 2, 3}
s.remove(2)
print(s)
```

### 14. symmetric_difference() (^)

```python
a = {1, 2, 3}
b = {3, 4}

print(a.symmetric_difference(b))
print(a ^ b)
```

### 15. symmetric_difference_update() (^=)

```python
a = {1, 2, 3}
b = {3, 4}

a.symmetric_difference_update(b)
print(a)
```

| Feature                      | `symmetric_difference()` | `symmetric_difference_update()` |
| ---------------------------- | ------------------------ | ------------------------------- |
| Returns a new set            | Yes                      | No                              |
| Modifies original set        | No                       | Yes                             |
| Original set after operation | Unchanged                | Changed                         |
| Can use operator             | `^`                      | `^=`                            |


### 16. union() (|)

```python
a = {1, 2, 5}
b = {3, 4, 5}

print(a.union(b))
print(a | b)
```

### 17. update() (|=)

```python
a = {1, 2}
b = {3, 4}

a.update(b)
print(a)
```
