### Frozenset
frozenset is an immutable version of a set.

Like sets, it contains unique, unordered, unchangeable elements.

Unlike sets, elements cannot be added or removed from a frozenset.


| Feature                                           | `set`                           | `frozenset`   |
| ------------------------------------------------- | ------------------------------- | ------------- |
| Type                                              | Mutable                         | **Immutable** |
| Can add elements                                  |  Yes (`add()`)                 |  No          |
| Can remove elements                               |  Yes (`remove()`, `discard()`) |  No          |
| Elements are unique                               |  Yes                           |  Yes         |
| Order of elements                                 | Unordered                       | Unordered     |
| Hashable                                          |  No                            |  **Yes**     |
| Can be used as dictionary key                     |  No                            |  Yes         |
| Can be used as set element                        |  No                            |  Yes         |
| Supports set operations (`union`, `intersection`) |  Yes                           |  Yes         |




### What does *hashable* mean?

**Hashable means “unchangeable”**, so Python can safely remember it.



### Why only some objects are hashable?

| Reason                | Simple Explanation                           |
| --------------------- | -------------------------------------------- |
| Value does not change | If the value changes, Python would “lose” it |
| Safe to remember      | Python stores a fixed code (hash) for it     |
| Can be found quickly  | Python uses the hash to find it fast         |



### Real-life example 

Think of **hash** like an **address**.

*  **Immutable object** → address never changes → **hashable**
*  **Mutable object** → address keeps changing → **not hashable**



### Why `frozenset` is hashable but `set` is not

| Object      | Reason (easy)                                        |
| ----------- | ---------------------------------------------------- |
| `frozenset` | You **cannot change** items → safe → hashable        |
| `set`       | You **can add/remove** items → unsafe → not hashable |


->

| Type        | Mutable | Hashable | Can be dict key |
| ----------- | ------- | -------- | --------------- |
| `set`       | Yes     |  No     |  No            |
| `frozenset` | No      |  Yes    |  Yes           |



### In One-line 

> An object is hashable because its value cannot change, so Python can store and find it easily.

### Examples of Hashable Objects
| Object      | Hashable | Reason                                   |
| ----------- | -------- | ---------------------------------------- |
| `int`       |  Yes    | Value never changes                      |
| `float`     |  Yes    | Immutable                                |
| `str`       |  Yes    | Immutable                                |
| `tuple`     |  Yes    | Immutable (if all elements are hashable) |
| `frozenset` |  Yes    | Immutable set                            |


### Examples of Non-Hashable Objects
| Object | Hashable | Reason  |
| ------ | -------- | ------- |
| `list` |  No     | Mutable |
| `set`  |  No     | Mutable |
| `dict` |  No     | Mutable |





### Creating a frozenset
Use the frozenset() constructor to create a frozenset from any iterable.

```python
x = frozenset({"apple", "banana", "cherry"})
print(x)
print(type(x))
```

### Frozenset Method
| Method                   | Shortcut   | Easy Description                                    |                                          
| ------------------------ | ---------- | --------------------------------------------------- | 
| `copy()`                 | —          | Creates a duplicate of the frozenset                |                                          
| `difference()`           | `-`        | Returns elements present in first but not in second |                                          
| `intersection()`         | `&`        | Returns common elements from both                   |                                          
| `isdisjoint()`           | —          | Checks if no elements are common                    |                                          
| `issubset()`             | `<=` / `<` | Checks if all elements exist in another frozenset   |                                          
| `issuperset()`           | `>=` / `>` | Checks if it contains all elements of another       |                                          
| `symmetric_difference()` | `^`        | Returns elements that are not common                |                                          
| `union()`                |           | Combines all elements without duplicates |


### 1. copy() method

```python
fs = frozenset({1, 2, 3})
cp = fs.copy()
print(cp)
```

### 2. difference() or (-)

```python
a = frozenset({1, 2, 3, 4, 6})
b = frozenset({3, 4, 5})
print(a.difference(b)) # or
print(a - b) 
```

### 3. intersection() or (&)

```python
a = frozenset({1, 2, 3, 4})
b = frozenset({3, 4, 5})
print(a.intersection(b)) # or
print(a & b)
```

### 4. isdisjoint() 

```python
a = frozenset({1, 2})
b = frozenset({3, 4})
c = frozenset({2, 3})
print(a.isdisjoint(b))
print(a.isdisjoint(c))
```

### 5. issubset() or <= or <

```python
a = frozenset({1, 2})
b = frozenset({1, 2, 3})
print(a.issubset(b)) # or
print(a <= b) 
print(a < b)
```

All elements of a are present in b.

### 6. issuperset() or >= or >

```python
a = frozenset({1, 2, 3})
b = frozenset({1, 2})
print(a.issuperset(b))
print(a >= b)
print(a > b)
```

All elements of b are present in a.

### 7. symmetric_difference or ^

```python
a = frozenset({1, 2, 3})
b = frozenset({3, 4, 5})
print(a.symmetric_difference(b)) # or
print(a ^ b)
```

### 8. union() or |

```python
a = frozenset({1, 2})
b = frozenset({2, 3})
print(a.union(b))
print(a | b)
```
