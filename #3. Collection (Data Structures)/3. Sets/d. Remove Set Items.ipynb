## Remove Set Items
To remove an item in a set, use the remove(), or the discard() method.

### 1. remove() method

```python
fruits = {"apple", "banana", "cherry"}
fruits.remove("banana")
print(fruits)
```

```python
even = {2, 4, 6, 8}
even.remove(6)
print(even)
```

```python
myset = {1, 2, 3, 4}
myset.remove(1)
myset.remove(2)
print(myset)
```

Note: If the item to remove does not exist, remove() will raise an error.

### 2. discard() method


```python
fruits = {"apple", "banana", "cherry"}
fruits.discard("banana")
print(fruits)
```

```python
numbers = {10, 20, 30}
numbers.discard(20)
print(numbers)
```

### Item not present

```python
fruits = {"apple", "banana"}
fruits.discard("mango")
print(fruits)
```

Note: If the item to remove does not exist, discard() will NOT raise an error.

### pop() method
You can also use the pop() method to remove an item, but this method will remove a random item, so you cannot be sure what item that gets removed.

Since sets are unordered, you cannot control which element is removed.

The return value of the pop() method is the removed item.



```python
fruits = {"apple", "banana", "cherry"}

item = fruits.pop()
print(item)
print(fruits)
```

```python
numbers = {1, 2, 3, 4}
print(numbers.pop())
print(numbers)
```

### clear() method
The clear() method empties the set

```python
fruits = {"apple", "banana", "cherry"}
fruits.clear()
print(fruits)
```

```python
colors = {"red", "blue", "green"}
colors.clear()
print(colors)
```

```python
marks = {70, 80, 90, 100}
marks.clear()
print(marks)
```
