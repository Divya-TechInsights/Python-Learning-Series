### Check Set Items
We cannot access items in a set by referring to an index or a key.

But we can ask if a specified value is present in a set, by using the in keyword.

### 1. Using in

```python
_set = {"apple", "banana", "cherry"}
print("banana" in _set)
```

```python
numbers = {10, 20, 30, 40}
print(20 in numbers)
print(50 in numbers)
```

```python
myset = set()

print("test" in myset)
print("example" in myset)
```

### 2. Using not in


```python
_set = {"apple", "banana", "cherry"}
print("banana" not in _set)
```

```python
numbers = {10, 20, 30, 40}
print(25 not in numbers)
print(50 not in numbers)
```

```python
myset = set()
print("number" not in myset)
```
