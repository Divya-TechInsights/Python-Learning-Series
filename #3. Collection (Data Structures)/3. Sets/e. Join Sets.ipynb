## Join Sets

| Method                   | What it Does                                 | Keeps                           |
| ------------------------ | -------------------------------------------- | ------------------------------- |
| `union()`                | Joins all items from sets                    | All items (no duplicates)       |
| `update()`               | Adds items from another set into current set | All items                       |
| `intersection()`         | Finds common items                           | **Only duplicates**             |
| `difference()`           | Finds unique items in first set              | Items **not** in other set      |
| `symmetric_difference()` | Finds non-common items                       | All items **except duplicates** |


### 1. union() – Join All Items

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
```

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1.union(set2)
print(set3)
```

#### Join multiple sets

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

myset = set1.union(set2, set3, set4)
print(myset)
```

 ### Use the | operator instead of the union() method, and you will get the same result

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1 | set2
print(set3)
```

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

myset = set1 | set2 | set3 |set4
print(myset)
```

### Join a Set and a Tuple
The union() method allows you to join a set with other data types, like lists or tuples.

The result will be a set.

```python
x = {"a", "b", "c"}
y = (1, 2, 3)

z = x.union(y)
print(z)
```

### Join Set with List

```python
x = {"a", "b", "c"}
y = [1, 2, 3]

z = x.union(y)
print(z)
```

### Join Set with String

```python
s = {"x", "y"}
st = "yz"

print(s.union(st))
```

### Join Set with Dictionary

```python
s = {1, 2}
d = {3: "a", 4: "b"}

print(s.union(d))
```

### Join Set with Multiple Iterables

```python
s = {1}
print(s.union([2, 3], (4, 5), {6}))
```

### 2. update() – Join and Modify Original Set

```python
A = {1, 2}
B = {2, 3, 4}

A.update(B)
print(A)
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

x.update(y)

print(x)
```

### Updating with multiple sets

```python
s = {10, 20}
s1 = {30, 40}
s2 = {50, 60}

s.update(s1, s2)
print(s)
```

### Use |= as a shortcut instead of update()

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

x |= y # x = x | y

print(x)
```

```python
a = {1, 2, 3}
b = {3, 4, 5}

a |= b
print(a)
```

```python
s = {10, 20}
s |= {30, 40}
s |= {20, 50}

print(s)
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}
z = {"cherry", "micra", "bluebird"}

x |= y | z

print(x)
```

### 3. intersection() Method
The intersection() method returns a new set containing only the common elements between the set and other iterable(s).

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.intersection(y)

print(z)
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}

print(a.intersection(b))
```

```python
x = {"a", "b", "c"}
y = {"c", "d", "e"}
z = {"f", "g", "c"}

result = x.intersection(y, z)

print(result)
```

### with List

```python
s = {1, 2, 3}
lst = [2, 3, 4]

print(s.intersection(lst))
```

### with Tuple

```python
s = {"a", "b", "c"}
t = ("b", "c", "d")

print(s.intersection(t))
```

### with a String

```python
s = {"a", "b", "c"}
st = "bcde"

print(s.intersection(st))
```

### with a dictionary

```python
s = {1, 2, 3}
d = {2: "x", 3: "y"}

print(s.intersection(d))
```

Note - Only dictionary keys are considered

### with multiple iterables

```python
s = {1, 2, 3, 4}
print(s.intersection([2, 3, 4], (3, 4), {4, 5}))
```

### Use & as a shortcut instead of intersection()

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x & y

print(z)
```

```python
x = {"a", "b", "c"}
y = {"c", "d", "e"}
z = {"f", "g", "c"}

result = x & y & z

print(result)
```

### 4. difference() Method
The difference() method returns a set that contains the difference between two sets.

Meaning: The returned set contains items that exist only in the first set, and not in both sets.

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.difference(y)

print(z)
```

```python
a = {"apple", "banana", "cherry"}
b = {"google", "microsoft", "apple"}
c = {"cherry", "micra", "bluebird"}

myset = a.difference(b, c)

print(myset)
```

### Use - as a shortcut instead of difference()

```python
a = {"apple", "banana", "cherry"}
b = {"google", "microsoft", "apple"}

myset = a - b

print(myset)
```

```python
a = {"apple", "banana", "cherry"}
b = {"google", "microsoft", "apple"}
c = {"cherry", "micra", "bluebird"}

myset = a - b - c

print(myset)
```

### Return a set that contains the items that only exist in set y, and not in set x

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = y.difference(x)

print(z)
```

### 5. symmetric_difference() Method
The symmetric_difference() method returns a set that contains all items from both set, but not the items that are present in both sets.

Meaning: The returned set contains a mix of items that are not present in both sets.

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.symmetric_difference(y)

print(z)
```

### with List

```python
s = {1, 2, 3}
lst = [3, 4, 5]

print(s.symmetric_difference(lst))
```

### with Tuple

```python
s = {"a", "b", "c"}
t = ("b", "d")

print(s.symmetric_difference(t))
```

### with String

```python
s = {"a", "b"}
st = "bc"

print(s.symmetric_difference(st))
```

### with a dictionary

```python
s = {1, 2, 3}
d = {3: "x", 4: "y"}

print(s.symmetric_difference(d))
```

### Use ^ as a shortcut instead of symmetric_difference()

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x ^ y

print(z)
```
