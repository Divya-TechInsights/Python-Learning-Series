### Add Items
Once a set is created, you cannot change its items, but you can add new items.

### 1. add() method
To add one item to a set use the add() method.

```python
_set = {"apple", "banana", "cherry"}

_set.add("orange")

print(_set)
```

### 2. Update() method
#### a. To add items from another set into the current set, use the update() method.

```python
_set = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}

_set.update(tropical)

print(_set)
```

#### b. Add Any Iterable
The object in the update() method does not have to be a set, it can be any iterable object (tuples, lists, dictionaries etc.).

```python
# Adding elements from a list to a set
_set = {"apple", "banana", "cherry"}
mylist = ["kiwi", "orange"] 

_set.update(mylist)

print(_set)
```

```python
# Adding elements from a tuple to a set
myset = {"a", "b"}
t = ("c", "d")
myset.update(t)
print(myset)
```

```python
# Adding elements from a dictionary to a set
myset = {1, 2}
myset.update({3: "three", 4: "four"})
print(myset)
```

```python
# Adding elements from a string to a set
myset = {"x"}
myset.update("yz")
print(myset)
```

| Iterable Type | Example Passed to `update()` | What Gets Added   |
| ------------- | ---------------------------- | ----------------- |
| List          | `[3, 4]`                     | Elements of list  |
| Tuple         | `("a", "b")`                 | Elements of tuple |
| Set           | `{5, 6}`                     | Elements of set   |
| Dictionary    | `{1:"a", 2:"b"}`             | Only keys         |
| String        | `"abc"`                      | Each character    |
