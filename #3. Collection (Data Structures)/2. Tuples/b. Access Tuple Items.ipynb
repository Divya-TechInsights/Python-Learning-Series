## Access Tuple Items

Tuples are ordered, so every item has an index (position).

### 1. Access by Index

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple[0])   # apple
print(thistuple[1])   # banana
print(thistuple[2])   # cherry
```

### 2. Negative Indexing

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple[-1])  # cherry
print(thistuple[-2])  # banana
print(thistuple[-3])  # apple
```

### 3. Range of Indexes (Slicing)

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi")
print(thistuple[1:4])   # ('banana', 'cherry', 'orange')
```

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[2:5])  # ('cherry', 'orange', 'kiwi')
```

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[:4])  # ('apple', 'banana', 'cherry', 'orange')
```

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[2:])  # ('cherry', 'orange', 'kiwi', 'melon', 'mango')
```
