### Tuple Methods

Python has two built-in methods that can be used on tuples.

### 1. count() Method
Returns how many times a value appears in a tuple.

```python
thistuple = ("apple", "banana", "cherry", "apple", "apple")

x = thistuple.count("apple")
print(x)
```

### 2. index() Method
Searches for a value and returns the index (position) of its first occurrence.

```python
thistuple = ("apple", "banana", "cherry")

x = thistuple.index("banana")
print(x)
```

### Combined Example

```python
numbers = (1, 2, 3, 2, 4, 2,1)

print(numbers.count(2))   # 3
print(numbers.index(2))   # 1
print(numbers.index(2,2)) # 3
```
