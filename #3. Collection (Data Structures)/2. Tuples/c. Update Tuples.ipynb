## Update Tuples

Tuples are immutable (unchangeable)

So we cannot change, add, or remove items.

Then it shows "workarounds".

### Meaning of “Workarounds”

Workarounds = Smart alternative ways to do something when the direct way is not possible

#### In tuple:

Direct way: Change tuple item (not allowed)

Workaround: Convert to list → change → convert back

### 1. Change Tuple Values

````python
x = ("apple", "banana", "cherry")
y = list(x) 
print(y)  # Step 1: Convert tuple → list
         
y[1] = "kiwi"  # Step 2: Modify list element
print(y)
    
x = tuple(y)       # Step 3: Convert list → tuple
print(x)
````

### Working of “Change Tuple Value” Code

```python
x = ("apple", "banana", "cherry")
```

Python creates a **tuple object in memory**.

Memory (concept):

```
x → ('apple', 'banana', 'cherry')
```

---

```python
y = list(x)
```

Python **copies all values** from tuple into a **new list object**.

```
x → ('apple', 'banana', 'cherry')
y → ['apple', 'banana', 'cherry']
```

---

```python
y[1] = "kiwi"
```

Now Python changes **only the list**, not the tuple.

```
y → ['apple', 'kiwi', 'cherry']
```

Tuple `x` is still untouched.

---

```python
x = tuple(y)
```

Python creates a **new tuple** from the list.









### 2. Add Items
#### a. Convert into a list

Workaround: Convert tuple → list → tuple

#### Why can’t we use append() on a tuple?

A tuple is immutable → once created, it cannot be changed

So methods like append(), remove(), insert() do not exist for tuples

That’s why we use a workaround.

````python
x = ("apple", "banana", "cherry")
y = list(x)
print(y) # Step 1: Convert tuple → list

y.append("orange") # Step 2: Add value to list
print(y)   
 
x = tuple(y)      # Step 3: Convert list → tuple
print(x)
````

### b. Add a tuple to a tuple

Even though tuples are immutable, we are allowed to combine (concatenate) tuples.

- We are not modifying the original tuple
- We are creating a new tuple

````python
thistuple = ("apple", "banana", "cherry")
y = ("orange",)
thistuple += y  #thistuple = thistuple + y

print(thistuple)
````
