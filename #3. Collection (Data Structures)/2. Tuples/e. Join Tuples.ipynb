## Join Tuples

### 1. Join Two Tuples
To join two or more tuples you can use the + operator

```python
tuple1 = ("a", "b" , "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```

### Multiply Tuples
To multiply the content of a tuple a given number of times, we can use the * operator

```python
fruits = ("apple", "banana", "cherry")
tuple1 = fruits * 2
print(tuple1)

tuple2 = fruits * 3
print(tuple2)
```
