### Packing a Tuple

When you put values into a tuple, it is called packing.

```python
fruits = ("apple", "banana", "cherry")
```

"apple", "banana", "cherry" are packed into one variable fruits

fruits is a tuple

### Unpacking a Tuple

When you take values out of a tuple and store them into variables, it is called unpacking.

```python
fruits = ("apple", "banana", "cherry")

(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)
```

### What Python does internally

green  ← "apple"

yellow ← "banana"

red    ← "cherry"

Note - Number of variables must match number of tuple items

### Using Asterisk * 

When the number of variables does not match, use * to collect extra values.

```python
fruits = ("apple", "banana", "cherry", "orange", "mango")

(a, b, *c) = fruits

print(a)
print(b)
print(c)
```

*c collects remaining items as a list, not a tuple.

```python
fruits = ("apple", "banana", "cherry", "orange")

(a, *b, c) = fruits

print(a)
print(b)
print(c)
```
