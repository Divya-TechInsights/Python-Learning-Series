## for Loop 

A for loop is used to go through (iterate over) each item in a sequence one by one.

A sequence can be:

- List
- Tuple
- String
- Set
- Dictionary

Instead of repeating code manually, we use a for loop.

#### Syntax:
```python
for variable in sequence:
    # code block
```

### 1. Loop Through a List

````python
fruits = ["apple", "banana", "mango"]

for x in fruits:
    print(x)
````

#### Explanation:

- First time → fruit = "apple"
- Second time → fruit = "banana"
- Third time → fruit = "mango"

### 2. Loop Through a String

````python
for letter in "Python":
    print(letter)
````

### 3. Loop Through a Tuple

````python
numbers = (10, 20, 30)

for num in numbers:
    print(num)
````

### 4. Loop Through a Set

````python
colors = {"red", "green", "blue"}

for color in colors:
    print(color)
````

### 5. Loop Through a Dictionary

````python
student = {"name": "Dev", "age": 22}

for key in student:
    print(key)
    print(student[key])

for key, value in student.items(): # .items() - returns a list of tuples containing key-value pairs from the dictionary
    print(key, value)
````

### The break Statement 

- The break statement is used to immediately stop the loop, even if there are items left.
- As soon as Python sees break, it exits the loop completely.

````python
fruits = ["apple", "banana", "cherry"]

for x in fruits:
    print(x)
    if x == "banana":
        break
````

#### Step-by-step Execution

1. x = "apple" → printed
2. x = "banana" → printed
- Condition becomes True → break runs
3. Loop stops 
4. "cherry" is never printed

````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    break
  print(x)
````

````python
numbers = [10, 20, 30, 40, 50]

for num in numbers:
    if num == 30:
        break
    print(num)
````

````python
name = "Kavya"

for letter in name:
    if letter == "v":
        print("Found!")
        break
````

### The continue Statement 

The continue statement is used to:

Skip the current iteration and move to the next iteration of the loop.

- It does NOT stop the loop completely.
- It only skips one round.

````python
fruits = ["apple", "banana", "cherry"]

for x in fruits:
    if x == "banana":
        continue
    print(x)
````

#### Step-by-step Execution
```python
x = "apple"
Condition False → print("apple")

x = "banana"
Condition True → continue runs
Print statement is skipped 

x = "cherry"
Condition False → print("cherry")
```

| Statement  | What It Does                 |
| ---------- | ---------------------------- |
| `break`    | Stops the entire loop        |
| `continue` | Skips current iteration only |


### The range() Function in Python

range() is used when you want to repeat something a specific number of times.

It generates a sequence of numbers.



````python
for x in range(6):
    print(x)
````

#### Why does it stop at 5?

Because:

range(6) means start from 0 and stop BEFORE 6.

Different Forms of range()

There are 3 ways to use it:

| Syntax                     | Meaning                                |
| -------------------------- | -------------------------------------- |
| `range(stop)`              | Start from 0, stop before `stop`       |
| `range(start, stop)`       | Start from `start`, stop before `stop` |
| `range(start, stop, step)` | Start from `start`, increase by `step` |


### 1. Start and Stop

````python
for x in range(2, 6):
    print(x)
````

Starts from 2, stops before 6.

### 2. Using Step

````python
for x in range(1, 10, 2):
    print(x)
````

### 3. Reverse Counting

````python
for x in range(5, 0, -1):
    print(x)
````

````python
for x in range(2, 30, 3):
  print(x)
````

### else in a for Loop 

In Python, the else block after a for loop runs only when the loop finishes normally (without break).

````python
for x in range(6):
    print(x)
else:
    print("Finally finished!")
````

````python
for x in range(6):
    if x == 3:
        break
    print(x)
else:
    print("Finally finished!")
````

#### This is commonly used in searching logic.

````python
numbers = [2, 4, 6, 8]

for num in numbers:
    if num == 5:
        print("Found")
        break
else:
    print("Not Found")
````

### Nested Loops 

#### A nested loop means:
A loop inside another loop.

- The outer loop runs first.
- For each iteration of the outer loop, the inner loop runs completely.

````python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
    for y in fruits:
        print(x, y)
````

### Step-by-Step Execution
1. Iteration 1 (Outer Loop → x = "red")

Inner loop runs fully:
```python
red apple
red banana
red cherry
```

2. Iteration 2 (Outer Loop → x = "big")

Inner loop runs fully again:
```python
big apple
big banana
big cherry
```
3. Iteration 3 (Outer Loop → x = "tasty")

Inner loop runs fully again:
```python
tasty apple
tasty banana
tasty cherry
```
### Key Concept 
```python
If:

- Outer loop runs m times
- Inner loop runs n times

Total executions = m × n
Here:

3 adjectives
3 fruits
→ 3 × 3 = 9 outputs
```
### This concept is extremely important later in:

- Pattern printing
- Matrix operations
- Machine Learning (feature combinations)
- Time complexity (O(n²))

### Visual Representation
```python

FOR each adjective
    FOR each fruit
        PRINT combination

Outer controls categories,
Inner controls combinations.
```

````python
adj = ["small", "fast", "smart"]
animals = ["dog", "cat", "fox"]

for a in adj:
    for b in animals:
        print(a, b)
````

````python
adj = ["hot", "cold", "rainy"]
days = ["Monday", "Tuesday"]

for a in adj:
    for d in days:
        print(a, d)
````

````python
adj = ["excellent", "average", "poor"]
students = ["Rahul", "Priya"]

for a in adj:
    for s in students:
        print(s, "is", a)
````
