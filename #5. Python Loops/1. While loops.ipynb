### Python has two primitive loop commands:

1. while loops
2. for loops

## The while Loop
A while loop keeps running as long as a condition is True.

- It checks the condition first.
- If the condition is True → it runs the block.
- If the condition becomes False → it stops.

````python
i = 1        # Step 1: initialize variable

while i < 6: # Step 2: condition
    print(i)
    i += 1   # Step 3: update variable
````

### Step by Step Execution
| Step | Value of i | Condition (i < 6) | Output |
| ---- | ---------- | ----------------- | ------ |
| 1    | 1          | True              | 1      |
| 2    | 2          | True              | 2      |
| 3    | 3          | True              | 3      |
| 4    | 4          | True              | 4      |
| 5    | 5          | True              | 5      |
| 6    | 6          | False             | Stops  |

#### Why is i += 1 Important?

If you remove:

``` python i += 1

Then i will always stay 1.
Condition i < 6 will always be True.
That creates an infinite loop (runs forever).
And that will freeze your program. 
```

### Important Rule 

#### A while loop needs 3 things:

```python
1️ Initialization (starting value)
2️ Condition (when to stop)
3️ Update (change the value)
```
If any one is missing → problem.

### Print Numbers 1 to 10

````python
i = 1

while i <= 10:
    print(i)
    i += 1
````

### Print Even Numbers from 2 to 10

````python
i = 2

while i <= 10:
    print(i)
    i += 2
````

### Print Numbers in Reverse (10 to 1)

````python
i = 10

while i >= 1:
    print(i)
    i -= 1
````

### Sum of First 5 Numbers

````python
i = 1
total = 0

while i <= 5:
    total += i #total = total + i
    i += 1 # i = i + 1

print("Sum is:", total)
````

### Step by Step
| Step | i (before check) | Condition (i ≤ 5) | total = total + i | i after increment | total (after step) |
| ---- | ---------------- | ----------------- | ----------------- | ----------------- | ------------------ |
| 1    | 1                | True              | 0 + 1 = 1         | 2                 | 1                  |
| 2    | 2                | True              | 1 + 2 = 3         | 3                 | 3                  |
| 3    | 3                | True              | 3 + 3 = 6         | 4                 | 6                  |
| 4    | 4                | True              | 6 + 4 = 10        | 5                 | 10                 |
| 5    | 5                | True              | 10 + 5 = 15       | 6                 | 15                 |
| 6    | 6                | False             | —                 | —                 | Loop Stops         |


### Multiplication Table of 3

````python
i = 1

while i <= 10:
    print("3 x", i, "=", 3 * i)
    i += 1
````

### Remove Letters One by One

````python
word = "PYTHON"

while word != "":
    print(word)
    word = word[:-1]
````

````python
word = "education"
index = 0
count = 0

while index < len(word):
    if word[index] in "aeiou":
        count += 1
    index += 1

print("Vowels:", count)
````

### The break Statement
With the break statement we can stop the loop even if the while condition is true

````python
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
````

| Step | i (before check) | Condition (i < 6) | Printed | Is i == 3? | break executed?  | i after increment |
| ---- | ---------------- | ----------------- | ------- | ---------- | ---------------- | ----------------- |
| 1    | 1                | True              | 1       | No         | No               | 2                 |
| 2    | 2                | True              | 2       | No         | No               | 3                 |
| 3    | 3                | True              | 3       | Yes        | Yes → Loop Stops | —                 |


### Stop When Number Is Divisible by 7

````python
i = 1

while i <= 20:
    if i % 7 == 0:
        print("First multiple of 7:", i)
        break
    i += 1
````

### Stop When Character Found

````python
word = "education"
index = 0

while index < len(word):
    if word[index] == "a":
        print("Found 'a' at index:", index)
        break
    index += 1
````

### Stop When Sum Exceeds 30

````python
total = 0
i = 1

while i <= 10:
    total += i
    
    if total > 30:
        print("Sum exceeded 30")
        break
    
    i += 1

print("Final total:", total)
````

### Stop When List Element Is Negative

````python
numbers = [5, 8, 12, -3, 7, 9]
index = 0

while index < len(numbers):
    if numbers[index] < 0:
        print("Negative number found:", numbers[index])
        break
    index += 1
````

### The continue Statement
Continue does NOT stop the loop.

It:

- Skips the remaining code in the current iteration
- Jumps to the next loop cycle

````python
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
````

### Step by Step
| Step | i (after increment) | Condition (i < 6) | Is i == 3? | continue?        | Printed    |
| ---- | ------------------- | ----------------- | ---------- | ---------------- | ---------- |
| 1    | 1                   | True              | No         | No               | 1          |
| 2    | 2                   | True              | No         | No               | 2          |
| 3    | 3                   | True              | Yes        | Yes → Skip print | —          |
| 4    | 4                   | True              | No         | No               | 4          |
| 5    | 5                   | True              | No         | No               | 5          |
| 6    | 6                   | False             | —          | —                | Loop Stops |


### Skip Even Numbers

````python
i = 0

while i < 10:
    i += 1
    
    if i % 2 == 0:
        continue
    
    print(i)
````

### Skip Vowels in a Word

````python
word = "education"
index = 0

while index < len(word):
    if word[index] in "aeiou":
        index += 1
        continue
    
    print(word[index])
    index += 1
````

### Skip Negative Numbers

````python
numbers = [4, -2, 7, -5, 9]
index = 0

while index < len(numbers):
    if numbers[index] < 0:
        index += 1
        continue
    
    print(numbers[index])
    index += 1
````

### The else Statement
With the else statement we can run a block of code once when the condition no longer is true

````python
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
````

#### Note: The else block will NOT be executed if the loop is stopped by a break statement.

````python
numbers = [2, 4, 6, 8]
index = 0

while index < len(numbers):
    if numbers[index] == 5:
        print("Found 5")
        break
    index += 1
else:
    print("5 not found")
````
