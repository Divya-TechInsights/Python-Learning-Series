## range() 

range() is a built-in function that generates a sequence of numbers.

It is mostly used in loops like for.

#### Basic Syntax
```python
range(start, stop, step)
start → where numbers begin (default = 0)
stop → where numbers end (not included)
step → gap between numbers (default = 1)
```

### 1. Only stop value

````python
for i in range(5):
    print(i)
````

### 2. Start and stop

````python
for i in range(2, 6):
    print(i)
````

### 3. With step

````python
for i in range(1, 10, 2):
    print(i)
````

### What does “immutable” mean here?
Once a range is created, you cannot change it.

range() does not store all numbers in memory.

- It generates numbers one by one when needed
- That’s why it is memory efficient

range() is an immutable sequence generator in Python that produces numbers on demand, mainly used in loops.

### Using List to Display Ranges
The range object is a data type that represents an immutable sequence of numbers, and it is not directly displayable.

Therefore, ranges are often converted to lists for display.

````python
print(list(range(5))) 
print(list(range(1, 6)))
print(list(range(5, 20, 3)))
````

### Key idea 
- range() → gives numbers one by one 
- list(range()) → stores all numbers at once

### Slicing a range

Just like lists or strings, you can slice a range to get part of it.

````python
r = range(10) 
print(r[2])   
print(r[:3])
````

### Explanation
r[:3]

- This is slicing
- It returns a subsequence (range object)

- Start = default 0
- Stop = 3 (excluded)

Output: range(0, 3)

### Membership Testing
Ranges support membership testing with the in operator.

````python
r = range(0, 10, 2)
print(6 in r)
print(7 in r)
````

The return value is True when the number is present in the range, and False when it is not.

### Length
Ranges support the len() function to get the number of elements in the range.

````python
r = range(10)
print(len(r))
````

````python
r = range(2, 8)
print(len(r))
````

````python
r = range(0, 10, 2)
print(len(r))
````

- len(range()) gives the count of elements, not the last value
