### Python If Statement
An if statement in Python is a way to make decisions in your code. It lets the program run a block of code only if a condition is true.

if condition:

    # code to run if condition is True

condition is an expression that evaluates to True or False.

The code inside the if statement must be indented (usually 4 spaces).

### If statement runs in case condition is True

````python
a = 10
b = 20

if b > a:
    print("b is greater than a")
````

### Key Points 

Decision making: Only runs code when a condition is True.

Indentation matters: Python uses whitespace instead of {}.

Can have multiple lines: All lines in the block must have the same indentation.

````python
age = 20
if age >= 18:
    print("You are an adult")
    print("You can vote")
````

#### if statement = "Check a condition. If it’s true"

### 1. Comparison Operators
Used to compare values.

````python
# Equal to
x = 5
if x == 5:
    print("x is 5")
````

````python
# Not equal to
x = 5
if x != 3:
    print("x is not 3")
````

````python
# Less than
x = 5
if x < 10:
    print("x is less than 10")
````

````python
# Less than or equal
x = 5
if x <= 5:
    print("x is less than or equal to 5")
````

````python
# Greater than
x = 5
if x > 2:
    print("x is greater than 2")
````

````python
# Greater than or equal
x = 5
if x >= 5:
    print("x is greater than or equal to 5")
````

### 2. Logical Operators

Used to combine multiple conditions.

````python
# and - both conditions must be True
age = 20
income = 40000
if age >= 18 and income >= 30000:
    print("Eligible")
````

````python
# or - at least one condition is True
age = 16
permission = True
if age >= 18 or permission:
    print("Can enter")
````

````python
# not - reverses the condition
x = 5
if not x > 10:
    print("x is not greater than 10")
````

### 3. Membership Operators

Used to check if a value exists in a sequence (list, string, tuple, etc.)

````python
# in - checks if value is present in a sequence
fruits = ['apple', 'banana']

if 'apple' in fruits:
    print("Apple is available")
````

````python
fruits = ['apple', 'banana']

if 'orange' in fruits:
    print("Orange is available")
````

````python
# not in - checks if value is NOT present in a sequence
fruits = ['apple', 'banana']

if 'orange' not in fruits:
    print("Orange is not available")
````

````python
fruits = ['apple', 'banana']
if 'apple' not in fruits:
    print("apple is not available")
````

### 4. Identity Operators

Check if two variables point to the same object in memory.

````python
# is - True if both refer to the same object
x = [1, 2]
y = x
z = [1, 2]

if x is y:
    print("x and y are the same object")
````

````python
# is not - True if they refer to different objects
x = [1, 2]
y = x
z = [1, 2]

if x is not z:
    print("x and z are different objects")
````

### 5. Boolean Operators
Any expression that evaluates to True or False can be used directly.

````python
if True:
    print("This runs")
````

````python
if False:
    print("This will not run")
````

Python has a set of **reserved keywords** that **cannot be used as variable names**, function names, or identifiers because they have special meaning in the language.

```

and        continue   for        lambda     try
```

### Notes:
