### Match Statement
match is Python’s pattern matching statement (introduced in Python 3.10).

Instead of writing many if-elif-else, you can use match statement.

### Syntax
```python
match expression:
  case x:
    code block
  case y:
    code block
  case z:
    code block
```

Python checks the expression once, then compares it with each case.

````python
day = 4

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
````

### Default Case ( _ )

The underscore _ works like else

````python
day = 4
match day:
  case 6:
    print("Today is Saturday")
  case 7:
    print("Today is Sunday")
  case _:
    print("Looking forward to the Weekend")
````

### Combine Values
Use the pipe character | as an or operator in the case evaluation to check for more than one value match in one case

````python
day = 4
match day:
  case 1 | 2 | 3 | 4 | 5:
    print("Today is a weekday")
  case 6 | 7:
    print("I love weekends!")
````

### Guard Conditions (Extra if)
You can add extra condition inside a case.

````python
month = 5
day = 4
match day:
  case 1 | 2 | 3 | 4 | 5 if month == 4:
    print("A weekday in April")
  case 1 | 2 | 3 | 4 | 5 if month == 5:
    print("A weekday in May")
  case _:
    print("No match")
````

#### Here Python checks:

- Does day match 1–5?
- Is month == 5?

Both must be true.

````python
num = 10

match num:
    case 5:
        print("Five")
    case 10:
        print("Ten")
    case _:
        print("Other number")
````

````python
color = "red"

match color:
    case "red":
        print("Stop")
    case "green":
        print("Go")
    case "yellow":
        print("Wait")
    case _:
        print("Invalid color")
````

### When Should You Use match?

#### Use match when:

- You are checking one variable against many values
- You want cleaner code than many elif
- You want pattern matching (advanced use cases later)
