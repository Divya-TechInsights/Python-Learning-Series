### Python Elif Statement
elif stands for “else if”.

It is used to check another condition only if the previous if condition was False.

if condition1:

    # runs if condition1 is True
elif condition2:

    # runs if condition1 is False AND condition2 is True

Python checks from top to bottom and stops at the first True condition.


```python
marks = 85

if marks >= 90:
    print("A Grade")
elif marks >= 80:
    print("B Grade")
```

#### Step-by-step:

Is 85 ≥ 90?  No

Is 85 ≥ 80?  Yes

##### So it prints:

**B Grade**

And stops.

#### Important Point

Only one block runs in an if–elif chain — the first condition that becomes True.

#### When to Use elif

##### Use it when:

- You have multiple possible cases
- Only one result should happen
- You are checking ranges or categories

Think of it like decision-making:

“If not this… then check this… if not that… then check next…”

### How Python Executes if–elif

Python checks conditions from top to bottom:

1. It checks the if
2. If false → checks first elif
3. If false → checks next elif
4. Stops immediately when it finds the first True
5. Skips the rest

##### Very important:

Even if 3 conditions are True, only the first True block runs.

### Example 1 – Equal Numbers

```python
a = 33
b = 33

if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
```

#### Step-by-step thinking:

Is b > a? → No

Is a == b? → Yes

Print → "a and b are equal"

Stop checking.

### Example 2 – Grading System

```python
score = 75

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
```

### Now think logically:

75 ≥ 90? ❌

75 ≥ 80? ❌

75 ≥ 70? ✅

#### It prints:

**Grade: C**


Even though 75 is also ≥ 60, Python never checks that because it already found a True condition.

This is why order matters.

### Example 3 – Age Categories

```python
age = 25

if age < 13:
    print("You are a child")
elif age < 20:
    print("You are a teenager")
elif age < 65:
    print("You are an adult")
elif age >= 65:
    print("You are a senior")
```

### Check carefully:

25 < 13? ❌

25 < 20? ❌

25 < 65? ✅

#### So it prints:
**You are an adult**

And stops.

### When you SHOULD Use elif

#### Use elif when:

- Conditions are mutually exclusive
- Only one result should happen
- You are checking ranges
- You are checking categories

#### Examples:

- Grades
- Age groups
- Menu options
- Day numbers

#### Note: 
- If only one answer should come out → use if–elif.
- If multiple things can happen together → use multiple if.

### Wrong Way (Using Multiple if)

```python
score = 95

if score >= 90:
    print("Grade A")

if score >= 80:
    print("Grade B")

if score >= 70:
    print("Grade C")
```

#### Why?

Because:

95 ≥ 90 ✅

95 ≥ 80 ✅

95 ≥ 70 ✅

All three conditions are True.

So Python runs all three blocks.

But logically, a student cannot have A, B, and C at the same time.

### Correct Way (Using elif)

```python
score = 95

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
```

### Now what happens?

95 ≥ 90 ✅

Python prints "Grade A"

Python stops checking further conditions.

Only one grade is printed.

### Can a student have more than one grade at the same time?

No.

#### A student can be:

A

OR B

OR C

Not all three together.

That means these conditions are mutually exclusive — only one should be true in meaning.
