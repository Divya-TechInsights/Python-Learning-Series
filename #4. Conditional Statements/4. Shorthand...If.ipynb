## Shorthand If

### 1️ One-Line if (Shorthand If)
If you have only one action to perform for a condition, you can write it on the same line as the if.

#### Syntax:
```python
if condition: action
```

````python
a = 5
b = 2
if a > b: print("a is greater than b")
````

Since there’s only one statement (print(...)), we don’t need multiple lines.

````python
num = 10
if num > 0: print("Positive number")
````

````python
n = 8
if n % 2 == 0: print("Even")
````

### 2️ One-Line if ... else (Ternary Operator)

If you have one action for if and one for else, you can write it in a single line using a conditional expression.

#### Syntax:
```python
action_if_true if condition else action_if_false
```

#### A ternary operator is a short way to write an if-else condition in one single line.
##### It’s called ternary because it works with three parts:

1. Condition
2. Value if condition is True
3. Value if condition is False

````python
a = 2
b = 330
print("A") if a > b else print("B")
````

```
This checks the condition a > b. If true → prints "A", else → prints "B".
```

````python
password = "admin123"
print("Access Granted") if password == "admin123" else print("Access Denied")
````

````python
x = 50
y = 100
print("x is bigger") if x > y else print("y is bigger")
````

### 3️ Assigning a Value Using if ... else

You can assign a variable based on a condition in one line.

#### Syntax:
```python
variable = value_if_true if condition else value_if_false
```

````python
a = 10
b = 20
bigger = a if a > b else b
print("Bigger is", bigger)
````

Here, bigger will store the larger of a and b.

````python
num = 7
result = "Even" if num % 2 == 0 else "Odd"
print("result is", result)
````

````python
marks = 35
status = "Pass" if marks >= 40 else "Fail"
print(status)
````

````python
age = 17
message = "Adult" if age >= 18 else "Minor"
print(message)
````

### 4️ Multiple Conditions on One Line
You can chain ternary operators for multiple outcomes, but keep it short to maintain readability.

````python
a = 330
b = 330
print("A") if a > b else print("=") if a == b else print("B")
````

### Logic:

- If a > b → prints "A"
- Else, if a == b → prints "="
- Else → prints "B"

````python
marks = 75
grade = "A" if marks >= 90 else "B" if marks >= 75 else "C"
print(grade)
````

````python
temp = 30
status = "Hot" if temp > 35 else "Warm" if temp > 20 else "Cold"
print(status)
````

````python
temperature = 25
is_raining = True

wheather = "Very Hot" if temperature > 30 else "Nice Weather" if temperature > 20 and not is_raining else "Stay Inside"
print(wheather)
````

### 5 Practical Uses
#### a) Maximum of two numbers

````python
x = 15
y = 20
max_value = x if x > y else y
print("Maximum value:", max_value)
````

#### b) Setting a default value

````python
username = ""
display_name = username if username else "Guest"
print("Welcome,", display_name)
````

### 6️ When to Use Shorthand If

- Condition and action are simple.
- Improves code readability.
- For quick assignments based on conditions.

Avoid shorthand if for complex logic or multiple statements; use normal if-else blocks for clarity.

### Important Points

Shorthand if is powerful.

But if you start writing things like this:
```python
result = "A" if x > 90 else "B" if x > 80 else "C" if x > 70 else "D"
```

It becomes messy.

Readable code > Short code.

Good programmers optimize clarity first, not length.
