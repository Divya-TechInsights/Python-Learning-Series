## Pass Statement

##### pass is a placeholder statement.

It tells Python:

##### “Do nothing here — but don’t throw an error.”

It is required because Python does not allow empty blocks.


#### Why is it needed?

Python expects at least one statement inside:

- if
- elif
- else
- for
- while
- def
- class

- If you leave them empty →  IndentationError
- So you use pass →  No error

### Example Without pass (Error)

```python
score = 85

#if score > 90:
    # Excellent
```

### Example With pass (Correct)

```python
score = 85

if score > 90:
    pass  # Placeholder

print("Score processed")
```

| Basis of Comparison          | Comment (`#`)     | `pass`             |
| ---------------------------- | ----------------- | ------------------ |
| What is it?                  | A note for humans | A Python statement |
| Executed by Python?          | No                | Yes                |
| Does it perform any action?  | No                | No                 |
| Considered as a statement?   | No                | Yes                |
| Can it replace empty block?  | No                | Yes                |
| Prevents `IndentationError`? | No                | Yes                |
| Main Purpose                 | Explain code      | Maintain structure |

That’s why a comment alone cannot replace pass.

### When Should You Use pass?

Use it when:

1. You’re designing structure first
2. Logic will be written later
3. A case requires no action
4. You want an empty function or class

### What Does “pass in Development” Mean?

When we build programs, we don’t always write everything at once.

Professional developers often:

- First design the structure
- Then fill in the logic later

pass helps us create the structure without errors.

```python
age = 16

if age < 18:
    pass  # TODO: Add underage logic later
else:
    print("Access granted")
```

#### What is happening?

- Condition is checked.
- If age is less than 18 → nothing happens (for now).
- If age is 18 or more → print message.

We haven’t written the logic for underage users yet.
But the program still runs correctly.

That is development planning.

### Why Is This Useful?

Imagine you are building a big system like:

- Login system
- Banking app
- Student management system

You first decide:

- What conditions are needed
- What functions are required
- What classes are required

Then later you implement details.

Without pass, Python would give errors while you're still designing.

#### Note - During development, we use pass as a temporary placeholder until we write the real logic.

### pass with Multiple Conditions

```python
value = 50

if value < 0:
  print("Negative value")
elif value == 0:
  pass # Zero case - no action needed
else:
  print("Positive value")
```
