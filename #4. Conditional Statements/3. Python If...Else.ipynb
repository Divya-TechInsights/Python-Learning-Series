## Python Else Statement

### What else Really Means

#### else means:

“If none of the above conditions worked, then do this.”

That’s it.

It is the default action.

It only runs when all previous if and elif conditions are False.

### How Python Reads an If–Elif–Else Chain

#### Python checks from top to bottom:

1. Check if
2. If False → check elif
3. If False → check next elif
4. If all are False → run else

#### Very important:

-> The moment one condition is True, Python stops checking further.

### Example 1 – Understanding the Flow Clearly

````python
a = 200
b = 33

if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")
````

#### Step-by-step thinking:

- Is b > a? → 33 > 200 →  False
- Is a == b? → 200 == 33 →  False
- Nothing matched → go to else

### Else Without Elif

````python
a = 200
b = 33

if b > a:
    print("b is greater than a")
else:
    print("b is not greater than a")
````

#### This is just a two-way decision:

- If condition True → do this
- If condition False → do that

No middle case.

### Why else Must Come Last

Because it is the final fallback.

#### You cannot do this:

```python
if condition:
    ...
else:
    ...
elif another_condition:    ERROR
```


### Why?

#### Because else means:

“We are done checking conditions.”

After that, Python cannot accept another condition.

### Even or Odd Example

````python
number = 7

if number % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
````

### Logic:
- % gives remainder
- If remainder is 0 → even
- Otherwise → odd

There are only two possibilities.

So else is perfect here.

### Temperature Example (Range Logic)

````python
temperature = 22

if temperature > 30:
    print("It's hot outside!")
elif temperature > 20:
    print("It's warm outside")
elif temperature > 10:
    print("It's cool outside")
else:
    print("It's cold outside!")
````

#### If temperature is 22:

- 30? ❌
- 20? ✔ (True)

It stops here.

Even though 22 is also greater than 10, Python does NOT check further.

That’s why order matters.

### Important

```python
if temp > 30:
    ...
if temp > 20:
    ...
if temp > 10:
    ...
```
This is wrong for classification.

#### Because if temp = 35:

All three are True.

So all three blocks run.

That breaks the logic.

When only one result should happen → use elif.

### Think of it like doors:

- if → First door
- elif → Second door
- elif → Third door
- else → Emergency exit

Only one door opens.

### 1️ Positive, Negative, or Zero

````python
num = -5

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
````

### 2️ Grade Classification

````python
marks = 82

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
````

### 3 Divisible by 3 and 5

````python
number = 15

if number % 3 == 0 and number % 5 == 0:
    print("Divisible by both")
elif number % 3 == 0:
    print("Divisible by 3")
elif number % 5 == 0:
    print("Divisible by 5")
else:
    print("Not divisible by 3 or 5")
````

### 4 Username Validation

````python
username = "John"

if username == "":
    print("Username cannot be empty")
elif len(username) < 5:
    print("Username too short")
else:
    print("Valid username")
````

### 5 Age Category

````python
age = 17

if age >= 60:
    print("Senior Citizen")
elif age >= 18:
    print("Adult")
elif age >= 13:
    print("Teenager")
else:
    print("Child")
````

### 6 Tricky One

````python
x = 10

if x > 5:
    print("A")
if x > 8:
    print("B")
else:
    print("C")
````

### This is NOT an elif chain
#### Step-by-step:

- x > 5 → True → prints A
- x > 8 → True → prints B
- The else belongs to the second if
- Since second if is True → else does NOT run

### 7 Logical Operator Practice

````python
temperature = 25
is_raining = True

if temperature > 30:
    print("Very Hot")
elif temperature > 20 and not is_raining:
    print("Nice Weather")
else:
    print("Stay Inside")
````
