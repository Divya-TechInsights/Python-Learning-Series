### 16. In Python, an identifier:

Must start with a letter (a–z, A–Z) or underscore _

Can contain letters, digits, and underscores

Cannot start with a digit

Cannot be a keyword (like if, for, class, etc.)

No spaces or special characters

### What are ASCII characters?

ASCII characters have Unicode values from 0 to 127, including:

English letters (A–Z, a–z)

Digits (0–9)

Common symbols (@ # $ %)

Space and control characters (like \n, \t)

❌ It does not include:

Accented letters (é, ñ)

Emojis 😊

Non-English scripts (ह, 中)




##  What is ASCII?

ASCII stands for
American Standard Code for Information Interchange

It is a system that assigns a numeric value to characters so computers can store and process text.



##  ASCII value range

ASCII values go from:


0  →  127


Total = 128 characters



##  ASCII categories

#### How partition("-") works

-The separator is "-"

-Python finds the first occurrence of "-"

-The string is split into 3 parts


| Part             | Value       |
| ---------------- | ----------- |
| Before separator | `"Python"`  |
| Separator        | `"-"`       |
| After separator  | `"is-easy"` |


#### Rule when separator is NOT found

(original_string, "", "")

Since the separator is not found, the entire string is returned as the first element and the other two elements are empty strings.

### split()

The split() method in Python is used to break a string into a list of substrings based on a separator.

string.split(separator, maxsplit)

separator → optional (default: whitespace)

maxsplit → optional (maximum number of splits)

```python
# 30. split()
text = "Python is very easy"
print(text.split()) # default separator is space
```

```python
text = "Python-i-s-easy"
print(text.split("-")) # separator is hyphen
```

#### rsplit() method

Splits a string from right to left.

string.rsplit(separator, maxsplit)

```python
text = "A-B-C-D"

print(text.rsplit("-"))
print(text.rsplit("-", 2)) # split at most 2 times from the right

# default separator is space
print("Python is very easy".rsplit())
```

| Feature           | `split()`    | `rsplit()`   |
| ----------------- | ------------ | ------------ |
| Direction         | Left → Right | Right → Left |
| Output type       | List         | List         |
| Default separator | Whitespace   | Whitespace   |
| `maxsplit` effect | From left    | From right   |

#### splitlines() method 

splitlines() splits a string at line boundaries and returns a list of lines.

It understands:

\n (newline)

\r

\r\n

- Syntax

string.splitlines(keepends=False)

- Parameters

keepends (optional)

False → line break removed (default)

True → line break kept

```python
# splitlines
text = "Hello World!\nWelcome to Python.\nEnjoy coding!"
lines = text.splitlines()
print(lines)

#Using keepends=True
lines_with_ends = text.splitlines(keepends=True)
print(lines_with_ends)

#Different line separators
text = "Line1\rLine2\r\nLine3\nLine4"
lines = text.splitlines()
print(lines)

#Empty string
empty_text = ""
lines = empty_text.splitlines()
print(lines)  # Output: []
```

| Feature                    | `split()`         | `splitlines()` |
| -------------------------- | ----------------- | -------------- |
| Splits by                  | Separator / space | Line breaks    |
| Returns                    | List              | List           |
| Handles `\n`, `\r`, `\r\n` | ❌                 | ✅              |
| Default use                | Words             | Lines          |

#### 34. startswith() method 

startswith() checks whether a string begins with a specified prefix.

Returns True or False

Case-sensitive

Does not modify the string


🔹 Syntax: string.startswith(prefix, start, end)

🔹 Parameters

prefix → string or tuple of strings

start (optional) → starting index

end (optional) → ending index

```python
text = "Python programming"

print(text.startswith("Python"))
print(text.startswith("program"))
print(text.startswith("p"))  # Case-sensitive check

print(text.startswith("program", 7))
print(text.startswith("Python", 0, 6))  # Check within a range

#using startswith with tuples
filename = "data.csv"

print(filename.startswith(("data", "info", "report")))
print(filename.startswith(("img", "photo", "pic")))
```

#### 35. swapcase() method in Python

swapcase() converts uppercase letters to lowercase and lowercase letters to uppercase.

Non-alphabet characters (numbers, symbols, spaces) are unchanged

Returns a new string (strings are immutable)

Syntax: string.swapcase()

```python
# 35. Swapcase
text = "Python Programming"
print(text.swapcase())

text = "Hello World! 123" #mixed case with punctuation and numbers
print(text.swapcase())

#all lowercase
print("hello world".swapcase())

#all uppercase
print("HELLO WORLD".swapcase())
```

```python
# 38. zfill
txt = "50"
x = txt.zfill(10)
print(x)

# Example with negative number
num = "-42"
y = num.zfill(6)
print(y)

# Example with decimal number
dec = "3.14"
z = dec.zfill(8)
print(z)

# Example with negative decimal number
neg_dec = "-3.14"
a = neg_dec.zfill(8)
print(a)

# Example with already filled string
filled = "000123"
w = filled.zfill(7)
print(w)

# Example with empty string
empty = ""
v = empty.zfill(5)
print(v)

# Example with long string
long_str = "1234567890"
u = long_str.zfill(5)
print(u)

# Example with whitespace
whitespace = " 42"
t = whitespace.zfill(6)
print(t)

# Example with special characters
special = "#42"
s = special.zfill(6)
print(s)

# Example with alphanumeric string
alphanumeric = "A42"
r = alphanumeric.zfill(6)
print(r)

# Example with alphabetic string
alphabetic = "Hello"
q = alphabetic.zfill(10)
print(q)

# Example with date string
date_str = "2023-7-4"
p = date_str.zfill(12)
print(p)

# Example with time string
time_str = "9:30AM"
o = time_str.zfill(10)
print(o)

# Example for date formatting
day = "7"
month = "9"
print(day.zfill(2))
print(month.zfill(2))

# Example with URL string
url_str = "example.com"
n = url_str.zfill(20)
print(n)
```

Length of "3.14" → 4

Required width → 8

Zeros added on the left → 8 - 4 = 4 zeros
