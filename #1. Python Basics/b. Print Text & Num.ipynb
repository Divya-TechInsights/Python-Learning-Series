# STATEMENTS IN PYTHON

```python
#String Data Type
print("Hello, I am a Python script!")
print('Hello, I am a Python script!')
print("""Hello, I am a Python script!""")
print('''Hello, I am a Python script!''')
print("'Hello, I am a Python script!'")
print('"Hello, I am a Python script!"')
```

```python
print("Hello, I am a Python script!")
print("H")
print("I like Python programming.")
```

```python
print("Hello, I am a Python script!"); print("H"); print("I like Python programming.")
```

```python
#print("Hello, I am a Python script!") print("H") print("I like Python programming.")
```

# Print without a newline

```python
print("Hello, I am a Python script!", end=' ')
print("H")
print("I like Python programming.")
```

```python
print("Hello, I am a Python script!", end=' ')
print("H", end=' ')
print("I like Python programming.")
```
