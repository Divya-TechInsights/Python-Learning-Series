Python has **built-in data types** that tell Python what kind of data a variable is storing. 

### 1. Numeric Types

Used to store numbers.

| Data Type | Description                           | Example      |
| --------- | ------------------------------------- | ------------ |
| `int`     | Whole numbers                         | `x = 10`     |
| `float`   | Decimal numbers                       | `y = 3.14`   |
| `complex` | Numbers with real and imaginary parts | `z = 2 + 3j` |

<br>
<br>

### 2. Text Type

Used to store text.

| Data Type | Description            | Example           |
| --------- | ---------------------- | ----------------- |
| `str`     | Sequence of characters | `name = "Python"` |

<br>
<br>

### 3. Sequence Types

Used to store multiple values in an ordered way.

| Data Type | Ordered | Changeable | Example           |
| --------- | ------- | ---------- | ----------------- |
| `list`    | Yes     | Yes        | `a = [1, 2, 3]`   |
| `tuple`   | Yes     | No         | `b = (1, 2, 3)`   |
| `range`   | Yes     | No         | `r = range(1, 5)` |

<br>
<br>

### 4. Set Types

Used to store unique values (no duplicates).

| Data Type   | Ordered | Changeable | Example                     |
| ----------- | ------- | ---------- | --------------------------- |
| `set`       | No      | Yes        | `s = {1, 2, 3}`             |
| `frozenset` | No      | No         | `fs = frozenset({1, 2, 3})` |

<br>
<br>

### 5. Mapping Type

Used to store data in **key–value pairs**.

| Data Type | Description            | Example                          |
| --------- | ---------------------- | -------------------------------- |
| `dict`    | Stores key–value pairs | `d = {"id": 1, "name": "Nisha"}` |

<br>
<br>

### 6. Boolean Type

Used for true/false values.

| Data Type | Description   | Example           |
| --------- | ------------- | ----------------- |
| `bool`    | True or False | `is_valid = True` |

<br>
<br>

### 7. Binary Types

Used to store binary data.

| Data Type    | Description           | Example                   |
| ------------ | --------------------- | ------------------------- |
| `bytes`      | Immutable binary data | `b = b"abc"`              |
| `bytearray`  | Mutable binary data   | `ba = bytearray(5)`       |
| `memoryview` | View of binary data   | `mv = memoryview(b"abc")` |

<br>
<br>

### 8. None Type

Represents no value.

| Data Type  | Description      | Example    |
| ---------- | ---------------- | ---------- |
| `NoneType` | Absence of value | `x = None` |
