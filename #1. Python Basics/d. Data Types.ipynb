# PYTHON DATATYPES

| Example Code                                   | Data Type    |
| ---------------------------------------------- | ------------ |
| `x = "Hello World"`                            | `str`        |
| `x = 20`                                       | `int`        |
| `x = 20.5`                                     | `float`      |
| `x = 1j`                                       | `complex`    |
| `x = ["apple", "banana", "cherry"]`            | `list`       |
| `x = ("apple", "banana", "cherry")`            | `tuple`      |
| `x = range(6)`                                 | `range`      |
| `x = {"name": "John", "age": 36}`              | `dict`       |
| `x = {"apple", "banana", "cherry"}`            | `set`        |
| `x = frozenset({"apple", "banana", "cherry"})` | `frozenset`  |
| `x = True`                                     | `bool`       |
| `x = b"Hello"`                                 | `bytes`      |
| `x = bytearray(5)`                             | `bytearray`  |
| `x = memoryview(bytes(5))`                     | `memoryview` |
| `x = None`                                     | `NoneType`   |


```python
# String
x = "Hello World"
print(x)

#display the data type of x:
print(type(x)) 
```

```python
#int
x = 20
print(x)
print(type(x))
```

```python
#float
x = 20.5
print(x)
print(type(x))
```

```python
#complex
x = 100j
print(x)
print(type(x))
```

```python
#list
x = ["apple", "banana", "cherry"]
print(x)
print(type(x))
```

```python
#tuple
x = ("apple", "banana", "cherry")
print(x)
print(type(x))
```

```python
#range
x = range(6)
print(x)
print(type(x))
```

```python
#dictionary
x = {"name" : "John", "age" : 36}
print(x)
print(type(x))
```

```python
#set
x = {"apple", "banana", "cherry"}
print(x)
print(type(x))
```

```python
#frozen set
x = frozenset({"apple", "banana", "cherry"})
print(x)
print(type(x))
```

```python
#bool
x = False
print(x)
print(type(x))
```

```python
#bytes
x = b"Hello"
print(x)
print(type(x))
```

```python
#byte array
x = bytearray(5)
print(x)
print(type(x))
```

```python
#memory view
x = memoryview(bytes(5))
print(x)
print(type(x))
```

```python
#None Type
x = None
print(x)
print(type(x))
```
