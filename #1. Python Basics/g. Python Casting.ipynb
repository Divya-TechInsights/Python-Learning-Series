### Python Casting

```python
# Integers
x = int(1)   
y = int(5.8) 
z = int("7") 
print(x)  
print(y)
print(z)
```

```python
# Floats
x = float(1)     
y = float(5.8)   
z = float("8")   
w = float("7.2")
print(x)  
print(y)
print(z)  
print(w)
```

```python
# Strings
x = str("s1") 
y = str(2)    
z = str(3.0)
print(x)  
print(y)
print(z)
```
