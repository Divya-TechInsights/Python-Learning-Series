### Python Numbers

There are three numeric types in Python:

1. **int** – Integer numbers 
2. **float** – Floating point numbers (decimals) 
3. **complex** – Complex numbers 


```python
#Example: 
x = 1     # int
y = 2.8   # float
z = 1j    # complex
print(type(x))  
print(type(y))  
print(type(z)) 
```

```python
### Example: Integers in Python

x = 1
y = 35656222554887711
z = -3255522

print(type(x))
print(type(y))
print(type(z))
```

```python
### Example: Floats in Python

x = 1.10
y = 1.0
z = -35.59

print(type(x))
print(type(y))
print(type(z))
```

```python
#Example: Floats with Scientific Notation in Python
#Float can also be written using scientific notation with `e` to indicate the power of 10.

x = 35e3
y = 12E4
z = -87.7e100

print(type(x))
print(type(y))
print(type(z))
```

```python
#Complex
x = 3j
y = 1+4j
z = -5j

print(type(x))
print(type(y))
print(type(z))
```
