### What is a Decorator 

- Decorators let you add extra behavior to a function, without changing the function's code.
- A decorator is a function that takes another function as input and returns a new function.
#### A decorator is just a function that:

- takes another function
- adds some extra behavior
- returns a new modified function

### A basic decorator that uppercases the return value of the decorated function.

````python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

print(myfunction())
````

##  Step-by-Step Flow

### 1. Function is created

- `myfunction()` is defined
(It returns `"Hello Sally"`)



### 2. Decorator is applied

```python
@changecase
```

- Python converts it to:

```python
myfunction = changecase(myfunction)
```



### 3. Inside decorator

- `func = myfunction` (original function)

- New function `myinner()` is created:

* Calls `func()`
* Converts output to uppercase



### 4. Function gets replaced

```python
myfunction = myinner
```

- Now `myfunction` is no longer original
- It is the **wrapper function**



### 5. Function call

```python
print(myfunction())
```

Actually runs:

```python
myinner()
```



### 6. Execution inside wrapper

```text
func() → "Hello Sally"
→ .upper()
→ "HELLO SALLY"
```



## Final Output

```text
HELLO SALLY
```



## One Line Summary 

“Decorator replaces the function with a new function that modifies its output”



##  Short Flow of program

```text
Define function
→ Apply decorator
→ Function passed to decorator
→ Wrapper created
→ Function replaced
→ Call runs wrapper
```




By placing @changecase directly above the function definition, the function myfunction is being "decorated" with the changecase function.

The function changecase is the decorator.

The function myfunction is the function that gets decorated.

### Multiple Decorator Calls
A decorator can be called multiple times. Just place the decorator above the function you want to decorate.

````python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

@changecase
def otherfunction():
  return "I am speed!"

print(myfunction())
print(otherfunction())
````

###  What is happening here?

 Same decorator (`changecase`) is used on **multiple functions**



###  What this code does

 It converts the output of **both functions to uppercase**



#### Output

```text
HELLO SALLY
I AM SPEED!
```

### 1. Decorator is defined once

```python

    return func().upper()

```

### 2. First function gets decorated

```

###  When functions are called

```python
myfunction() → "HELLO SALLY"

```

#### Both go through:

- wrapper (`myinner`)

### Uppercase + Add text

````python
def uppercase(func):
    def wrapper():
        return func().upper()
    return wrapper

def add_hello(func):
    def wrapper():
        return "Hello " + func()
    return wrapper

@uppercase
@add_hello
def name():
    return "divya"

print(name())
````

### What is *args and **kwargs 

They allow a function to take any number of arguments

- *args → multiple positional arguments (like 1, 2, 3)
- **kwargs → multiple named arguments (like name="John")

### Why we use it in decorators?

Because we don’t know how many arguments the function will have

````python
def changecase(func):
  def myinner(*args, **kwargs):
    return func(*args, **kwargs).upper()
  return myinner

@changecase
def myfunction(nam):
  return "Hello " + nam

print(myfunction("John"))
````

### Execution Flow
```python
myfunction("John")
→ myinner("John")
→ func("John") → "Hello John"
→ .upper()
→ "HELLO JOHN"

### Print all arguments

````python
def show_args(func):
    def wrapper(*args, **kwargs):
        print("Args:", args)
        print("Kwargs:", kwargs)
        return func(*args, **kwargs)
    return wrapper

@show_args
def add(a, b):
    return a + b

print(add(2, 3))
````

### Works with keyword arguments

````python
def info(func):
    def wrapper(*args, **kwargs):
        print("User Info:", kwargs)
        return func(*args, **kwargs)
    return wrapper

@info
def person(name, age):
    return f"{name} is {age}"

print(person(name="Neha", age=21))
````

### Check condition

````python
def check_positive(func):
    def wrapper(*args, **kwargs):
        if args[0] < 0:
            return "Negative not allowed"
        return func(*args, **kwargs)
    return wrapper

@check_positive
def square(n):
    return n * n

print(square(5))
print(square(-3))
````

### Decorator With Arguments
Decorators can accept their own arguments by adding another wrapper level.


### What it means 

- Normally:

Decorator only takes a function

- But sometimes:

We want decorator to take its own value (like 1, 3, "Hi")

- For that, we need one extra function layer


### So we add “another wrapper level”

Structure becomes:

- Outer function → takes argument
- Inner function → takes function
- Wrapper → runs function

````python
def changecase(n):
  def changecase(func):
    def myinner():
      if n == 1:
        a = func().lower()
      else:
        a = func().upper()
      return a
    return myinner
  return changecase

@changecase(1)
def myfunction():
  return "Hello Linus"

print(myfunction())
````

### Repeat function n times

````python
def repeat(n):
    def decorator(func):
        def wrapper():
            for i in range(n):
                func()
        return wrapper
    return decorator

@repeat(3)
def hello():
    print("Hi")

hello()
````

Note: Decorator controls how many times function runs

### Add delay (real-world concept)

````python
import time

def delay(seconds):
    def decorator(func):
        def wrapper():
            time.sleep(seconds)
            return func()
        return wrapper
    return decorator

@delay(1)
def greet():
    print("Hello!")

greet()
````

### Logging with custom message

````python
def log(message):
    def decorator(func):
        def wrapper():
            print(message)
            return func()
        return wrapper
    return decorator

@log("Function is starting...")
def task():
    print("Working...")

task()
````

````python
def myfunction():
  return "Have a great day!"

print(myfunction.__name__)
````

````python
def myfunction():
    """This is a sample function"""
    return "Hello"

print(myfunction.__name__)
print(myfunction.__doc__)
````

### Important Note (connect to decorators)

- Without decorators:

metadata is correct

- With decorators:

metadata gets lost 

````python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

print(myfunction.__name__)
````

```python
Original → myfunction
↓ decorator applied
Replaced → myinner
↓
myfunction.__name__ → "myinner"
```

#### To fix this, Python has a built-in function called functools.wraps that can be used to preserve the original function's name and docstring.

Import functools.wraps to preserve the original function name and docstring.

````python
import functools

def changecase(func):
  @functools.wraps(func)
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

print(myfunction.__name__)
````
