## Python Generators
### What is a Generator?

A generator is a function that:

- returns values one at a time
- pauses execution using yield
- resumes from where it left off

Instead of returning all values at once (like a list), it gives them on demand.

````python
def my_generator():
  yield 1
  yield 2
  yield 3

for value in my_generator():
  print(value)
````

```python
Step-by-step:

my_generator() is called
It does NOT run immediately
It returns a generator object

for loop starts iterating
Now the function starts executing

Execution flow:
yield 1 → pauses → prints 1
resumes → yield 2 → prints 2
resumes → yield 3 → prints 3
function ends
```

| Feature   | `return`             | `yield`                   |
| --------- | -------------------- | ------------------------- |
| Returns   | Whole result at once | One value at a time       |
| Memory    | High (stores all)    | Low (lazy evaluation)     |
| Execution | Ends function        | Pauses & resumes function |


### Why Use Generators?

- Working with large data (files, streams)
- You don’t want to store everything in memory
- You want efficient iteration

### Without Generator

````python
def get_numbers():
    return [1, 2, 3]
````

Stores full list in memory

### With Generator

````python
def get_numbers():
    yield 1
    yield 2
    yield 3
````

Generates values one by one (more efficient)

### Important Insight 
A generator remembers its state.

````python
g = my_generator()

print(next(g))  # 1
print(next(g))  # 2
print(next(g))  # 3
````

### next() is a built-in function that:

- asks the generator for the next value
- resumes execution from where it last stopped

It doesn’t restart — it continues from where it paused.

Note - A generator is basically a lazy iterator with memory of execution state.

````python
def count_up_to(n):
  count = 1
  while count <= n:
    yield count
    count += 1

for num in count_up_to(5):
  print(num)
````

```python
Step-by-Step Execution

Step 1: count_up_to(5)
Generator object is created

Step 2: for loop starts
-> First iteration
Function starts
count = 1
yield count → returns 1
PAUSE here

-> Second iteration
Resumes from paused point
count += 1 → 2
yield 2
PAUSE again

-> Continues like this:
3
4
5

-> After 5
count = 6
Condition fails (6 <= 5)
Generator ends → StopIteration
```

### Generators Saves Memory

````python
def large_sequence(n):
  for i in range(n):
    yield i

# This doesn't create a million numbers in memory
gen = large_sequence(1000000)
print(next(gen))
print(next(gen))
print(next(gen))
````

* First `next(gen)` → gives **0**
* Second `next(gen)` → gives **1**
* Third `next(gen)` → gives **2**

Each time:

* The function **resumes from where it stopped**
* Runs until the next `yield`
* Returns one value and pauses again

So:

* `yield` = “give one value and pause”
* `next()` = “continue and give next value”

That’s why you need to call `next(gen)` again and again to get more values.


### Using next() with Generators

````python
def simple_gen():
  yield "Emil"
  yield "Tobias"
  yield "Linus"

gen = simple_gen()
print(next(gen))
print(next(gen))
print(next(gen))
````

#### When there are no more values to yield, the generator raises a StopIteration exception

````python
def simple_gen():
  yield 1
  yield 2

gen = simple_gen()
print(next(gen))
print(next(gen))
#print(next(gen)) # This will raise StopIteration
````

````python
# List comprehension - creates a list
list_comp = [x * x for x in range(5)]
print(list_comp)

# Generator expression - creates a generator
gen_exp = (x * x for x in range(5))
print(gen_exp)
print(list(gen_exp))
````

### Using a generator expression with sum

````python
# Calculate sum of squares without creating a list
total = sum(x * x for x in range(10))
print(total)
````

```python 
Python does this internally:

Generates 0², gives to sum()
Generates 1², gives to sum()
Generates 2², gives to sum()
… one by one
```

Because we used ( ) instead of [ ], it becomes a generator expression, so values are generated one by one instead of stored.

````python
# Calculate sum of squares without creating a list
total = sum([x * x for x in range(10)])
print(total)
````

- First creates full list → [0, 1, 4, 9, ..., 81]
- Then sum() adds all values

### Fibonacci Sequence Generator

````python
def fibonacci():
  a, b = 0, 1
  while True:
    yield a
    a, b = b, a + b

# Get first 100 Fibonacci numbers
gen = fibonacci()
for _ in range(100):
  print(next(gen))
````

### Each Step 

1. `a=0, b=1` → start values
2. `yield a` → output current number (0)
3. `a, b = b, a + b` → update (1,1)
4. `yield a` → output 1
5. update → (1,2)
6. `yield a` → output 1
7. update → (2,3)
8. `yield a` → output 2

   … continues



### Output flow

0 → 1 → 1 → 2 → 3 → 5 → …


## Generator Methods
Generators have special methods for advanced control:

### send() Method
send(value) sends a value into the generator

- The value goes inside yield
- It also moves the generator forward (like next())

````python
def echo_generator():
  while True:
    received = yield
    print("Received:", received)

gen = echo_generator()
next(gen) # Prime the generator
gen.send("Hello")
gen.send("World")
````

## Step-by-step 

### 1️. Create generator

```python
gen = echo_generator()
```

* Generator is created, **not started yet**



### 2️. Prime generator

```python
next(gen)
```

* Runs till `yield`
* Pauses at: `received = yield`
* Waiting to receive a value



### 3️. Send "Hello"

```python
gen.send("Hello")
```

* `"Hello"` goes into `yield`
* So: `received = "Hello"`
* Prints → **Received: Hello**
* Pauses again at `yield`



### 4️. Send "World"

```python
gen.send("World")
```

* `"World"` goes into `yield`
* Prints → **Received: World**
* Pauses again



### One-line

`send()` sends value into `yield`, prints it, and pauses again


### close() Method
The close() method stops the generator

````python
def my_gen():
  try:
    yield 1
    yield 2
    yield 3
  finally:
    print("Generator closed")

gen = my_gen()
print(next(gen))
gen.close()
````

## Step-by-step 

### 1️. Create generator

```python
gen = my_gen()
```

* Generator created (not started)



### 2️. First value

```python
print(next(gen))
```

* Runs till first `yield`
* Output: **1**



### 3️. Close generator

```python
gen.close()
```

* Stops generator immediately
* Executes `finally` block
* Output: **Generator closed**


## One line

- *`close()` ends the generator and executes the finally block*
- *close() kills the generator before it can give next values*
