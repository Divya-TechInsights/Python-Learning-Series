## Python Scope
### 1. What is Scope in Python?

Scope means the region of the program where a variable is accessible.

#### In simple words:
Where can we use the variable?

````python
def myfunc():
    x = 300
    print(x)

myfunc()
````

#### Explanation:

- x is created inside the function
- It can be used only inside the function
- If we try to print x outside → error

### 2. Local Scope

A variable created inside a function is called a local variable.

It works only inside that function.

````python
def myfunc():
    x = 'string'
    print(x)

myfunc()
````

### 3. Function Inside Function (Nested Functions)

A variable defined in the outer function can be used by the inner function.

````python
def myfunc():
    x = 300
    
    def myinnerfunc():
        print(x)
        
    myinnerfunc()

myfunc()
````

```python
Outer Function
   x = 300
      ↓
Inner Function
   can access x
   ```

### 4. Global Scope

A variable created outside all functions is called a global variable.

It can be used anywhere in the program.

````python
x = 300

def myfunc():
    print(x)

myfunc()
print(x)
````

x is accessible inside and outside the function.

### 5. Same Variable Name (Local vs Global)

If the same variable name is used inside and outside a function, Python treats them as different variables.

````python
x = 300

def myfunc():
    x = 200
    print(x)

myfunc()
print(x)
````

| Variable             | Scope  | Value |
| -------------------- | ------ | ----- |
| `x` inside function  | Local  | 200   |
| `x` outside function | Global | 300   |


### 6. Global Keyword

Sometimes we want to modify a global variable inside a function.

For that we use global keyword.

````python
def myfunc():
    global x
    x = 300

myfunc()
print(x)
````

Here x becomes global even though it was created inside the function.

### Modifying Global Variable

````python
x = 300

def myfunc():
    global x
    x = 200

myfunc()
print(x)
````

Function changed the global variable.

### 7. Nonlocal Keyword

nonlocal is used in nested functions to modify the outer function variable.

````python
def myfunc1():
    x = "Jane"
    
    def myfunc2():
        nonlocal x
        x = "hello"
        
    myfunc2()
    return x

print(myfunc1())
````

- Outer function → x = Jane
- Inner function → changes x to hello

### 8. LEGB Rule (How Python Finds Variables)

Python searches variables in this order:

L → E → G → B

| Level | Meaning                    |
| ----- | -------------------------- |
| **L** | Local (current function)   |
| **E** | Enclosing (outer function) |
| **G** | Global                     |
| **B** | Built-in                   |


````python
x = "global"

def outer():
    x = "enclosing"
    
    def inner():
        x = "local"
        print("Inner:", x)
        
    inner()
    print("Outer:", x)

outer()
print("Global:", x)
````

- Inner function → uses local x
- Outer function → uses enclosing x
- Main program → uses global x

### Python Variable Scope
- Local Scope      → inside function
- Enclosing Scope  → outer function
- Global Scope     → outside all functions
- Built-in Scope   → Python predefined names

### Search Order:
- LEGB
- Local → Enclosing → Global → Built-in
