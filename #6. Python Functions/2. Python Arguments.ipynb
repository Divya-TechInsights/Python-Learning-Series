### Python Function Arguments 
### 1. What are Arguments?

Arguments are values we pass to a function when we call it so the function can work with that data.

Think of a function like a machine.

- The input you give → Argument
- The variable that receives it → Parameter

````python
def greet(name):
    print("Hello", name)

greet("Divya")
````

#### Explanation:

- name → Parameter
- "Divya" → Argument

So,

- Parameter = variable in function definition
- Argument = value passed when calling the function

### 2. Multiple Arguments

A function can take more than one argument.

````python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Rahul", 20)
````

### Explanation:

- name → Rahul
- age → 20

The number of arguments must match the parameters.

### 3. Error if Arguments are Missing

````python
#def student(name, age):
#    print(name, age)

#student("Rahul")
````

Because 2 parameters exist but only 1 argument was given.

### 4. Default Parameter Values

We can give a default value to a parameter.

If the user does not pass an argument, the default value is used.

````python
def greet(name="Student"):
    print("Hello", name)

greet("Rahul")
greet()
````

#### Explanation:

- If name is provided → it uses that value
- If not → it uses "Student"

### 5. Keyword Arguments

Here we mention the parameter name while calling the function.

````python
def pet(animal, name):
    print("Animal:", animal)
    print("Name:", name)

pet(animal="Dog", name="Buddy")
pet(name="Buddy", animal="Dog")
````

#### Advantage:
- Order does not matter
- Same Output

### 6. Positional Arguments

When arguments are passed without specifying parameter names, they are positional.

Order matters.

````python
def pet(animal, name):
    print(animal, name)

pet("Dog", "Buddy")
pet('buddy', 'dog')
````

### If we switch order:
```python
pet("Buddy", "Dog")

Output
Buddy Dog
```
Wrong meaning because order changed.

### 7. Mixing Positional and Keyword Arguments

You can mix both.

But positional must come first.

````python
def pet(animal, name, age):
    print(animal, name, age)

pet("Dog", name="Buddy", age=5)
````

#### But this is invalid:
```python
pet(animal="Dog", "Buddy", age=5)
```

### 8. Passing Different Data Types

Functions can accept any data type.

````python
def show_fruits(fruits):
    for f in fruits:
        print(f)

fruit_list = ["Apple", "Banana", "Mango"]
show_fruits(fruit_list)
````

````python
def show_person(person):
    print("Name:", person["name"])
    print("Age:", person["age"])

p = {"name":"Rahul","age":22}

show_person(p)
````

### 9. Return Values

Functions can send results back using return.

````python
def add(x, y):
    return x + y

result = add(5,3)
print(result)
````

#### Explanation:

- Function calculates 5 + 3
- return sends result back
- Stored in result

### 10. Returning Multiple Values

````python
def numbers():
    return (10,20)

x,y = numbers()

print(x)
print(y)
````

### 11. Positional-Only Arguments (/)

Arguments before / must be positional.

````python
def greet(name, /):
    print("Hello", name)

greet("Rahul")
````

#### But this gives error:
```python
greet(name="Rahul")
```

### 12. Keyword-Only Arguments (*)

Arguments after * must be keyword arguments.

````python
def greet(*, name):
    print("Hello", name)

greet(name="Rahul")
````

### This is wrong:
```python
greet("Rahul")
```

### 13. Combining Both

````python
def add(a, b, /, *, c, d):
    return a + b + c + d

print(add(5,10,c=15,d=20))
````

| Type            | Parameters |
| --------------- | ---------- |
| Positional Only | a, b       |
| Keyword Only    | c, d       |


### Summary
| Type            | Meaning                         | Example         |
| --------------- | ------------------------------- | --------------- |
| Parameter       | Variable in function definition | `def fun(x)`    |
| Argument        | Value passed to function        | `fun(10)`       |
| Positional      | Order matters                   | `fun(5,10)`     |
| Keyword         | Use `key=value`                 | `fun(x=5,y=10)` |
| Default         | Value used if none given        | `def fun(x=10)` |
| Positional-only | Only positional allowed         | `def fun(x, /)` |
| Keyword-only    | Only keyword allowed            | `def fun(*, x)` |
