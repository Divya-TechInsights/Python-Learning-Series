## Python *Args and **Kwargs
1. Normal arguments
2. *args → many positional arguments
3. **kwargs → many keyword arguments

### 1. Problem First (Why *args and **kwargs?)

Normally a function needs a fixed number of arguments.

````python
def add(a, b):
    print(a + b)

add(5, 10)
````

#### But what if the user gives 3 numbers?
```python
add(5, 10, 15)    Error
```

#### Because the function expects only 2 arguments.

So Python gives a solution:

- *args
- **kwargs

These allow functions to accept any number of inputs.

### 2. *args (Arbitrary Positional Arguments)

*args allows a function to take any number of positional arguments.

```python
def function_name(*args):
```
Inside the function, args becomes a tuple.

````python
def my_function(*args):
    print(args)

my_function(10, 20, 30)
````

### Access Elements

````python
def my_function(*args):   
    print(args[0])
    print(args[1])

my_function("Apple", "Banana", "Mango")
````

### Loop through *args

````python
def show_numbers(*nums):
    for num in nums:
        print(num)

show_numbers(5,10,15,20)
````

### Sum Function

````python
def add_numbers(*numbers):
    total = 0

    for num in numbers:
        total = total + num

    return total

print(add_numbers(10,20))
print(add_numbers(10,20,30))
print(add_numbers(5))
````

#### Here the function works for any number of inputs.

### Step 1: Function Definition

```python
def add_numbers(*numbers):
```

* `add_numbers` → function name
* `*numbers` → allows the function to accept **any number of arguments**

All the values passed to the function are stored in **numbers as a tuple**.

Example:

```python
add_numbers(10,20,30)
```

Inside the function:

```
numbers = (10, 20, 30)
type(numbers) = tuple
```



### Step 2: Initialize Variable

```python
total = 0
```

We create a variable `total` and start it from **0**.

This variable will store the **sum of all numbers**.



### Step 3: Loop Through Numbers

```python
for num in numbers:
```

Since `numbers` is a tuple like:

```
(10, 20, 30)
```

The loop will run **once for each value**.

Iteration process:

| Iteration | num value |
| --------- | --------- |
| 1         | 10        |
| 2         | 20        |
| 3         | 30        |



### Step 4: Add Numbers

```python
total = total + num
```

Each number is added to `total`.

Example when calling:

```python
add_numbers(10,20,30)
```

Step-by-step calculation:

| Step  | total | num | total + num |
| ----- | ----- | --- | ----------- |
| Start | 0     | -   | 0           |
| 1     | 0     | 10  | 10          |
| 2     | 10    | 20  | 30          |
| 3     | 30    | 30  | 60          |

Final value:

```
total = 60
```



### Step 5: Return Result

```python
return total
```

The function sends the **final sum** back to the place where the function was called.



### Step 6: First Function Call

```python
print(add_numbers(10,20))
```

Inside function:

```
numbers = (10,20)
```

Calculation:

```
0 + 10 = 10
10 + 20 = 30
```

Return:

```
30
```

Output

```
30
```



### Step 7: Second Function Call

```python
print(add_numbers(10,20,30))
```

Inside function:

```
numbers = (10,20,30)
```

Calculation:

```
0 + 10 = 10
10 + 20 = 30
30 + 30 = 60
```

Output

```
60
```



### Step 8: Third Function Call

```python
print(add_numbers(5))
```

Inside function:

```
numbers = (5,)
```

Note: Single value tuple has **comma**.

Calculation:

```
0 + 5 = 5
```

Output

```
5
```



### Final Output

```
30
60
5
```






### 4. Regular Argument + *args
Normal arguments must come BEFORE *args

````python
def greet(greeting, *names):
    for name in names:
        print(greeting, name)

greet("Hello", "Aman", "Riya", "Rahul")
````

#### Explanation:
```python
greeting = "Hello"
names = ("Aman","Riya","Rahul")
```

### 5. **kwargs (Arbitrary Keyword Arguments)

**kwargs allows a function to accept any number of keyword arguments.
```python
def function_name(**kwargs):
```
Inside the function, kwargs becomes a dictionary.

````python
def my_function(**kwargs):
    print(kwargs)

my_function(name="Aman", age=25, city="Delhi")
````

### Accessing values

````python
def my_function(**data):
    print(data["name"])
    print(data["age"])

my_function(name="Aman", age=25)
````

### Looping through **kwargs

````python
def show_details(**info):

    for key, value in info.items():
        print(key, ":", value)

show_details(name="Aman", age=25, city="Delhi")
````
