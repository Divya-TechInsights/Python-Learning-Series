### Python Functions
#### 1. What is a Function?

A function is a block of code that runs only when it is called.

#### Functions help to:

- Avoid repeating the same code
- Organize programs
- Make code reusable

#### Basic Syntax
```python
def function_name():
    # code block
    

def → keyword used to create a function.

### 2. Creating a Function

````python
def my_function():
    print("Hello from a function")
````

### Explanation:

| Part          | Meaning                           |
| ------------- | --------------------------------- |
| `def`         | Keyword used to define a function |
| `my_function` | Function name                     |
| `()`          | Parentheses (used for parameters) |
| `:`           | Start of function block           |
| Indentation   | Code inside the function          |

Important: Python uses indentation to define code blocks.

### 3. Calling a Function

To run a function, write its name followed by ().

````python
def my_function():
    print("Hello from a function")

my_function()
````

### 4. Calling Function Multiple Times

````python
def my_function():
    print("Hello from a function")

my_function()
my_function()
my_function()
````

The same function can be used many times.

### 5. Rules for Function Names

Function names follow the same rules as variables.

| Rule                              | Example           |
| --------------------------------- | ----------------- |
| Must start with letter or `_`     | `_calculate()`    |
| Can contain letters, numbers, `_` | `sum1()`          |
| Cannot start with number          | `1sum()`        |
| Case sensitive                    | `sum()` ≠ `Sum()` |


#### Examples of valid names:
- calculate_sum()
- _private_function()
- myFunction2()
- calculate_average()
- convert_temperature()


### 6. Why Use Functions?

Without functions → code repetition

````python
temp1 = 77
celsius1 = (temp1 - 32) * 5 / 9
print(celsius1)

temp2 = 95
celsius2 = (temp2 - 32) * 5 / 9
print(celsius2)

temp3 = 50
celsius3 = (temp3 - 32) * 5 / 9
print(celsius3)
````

#### Problem:

Same formula repeated many times.

### Using Functions

````python
def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

print(fahrenheit_to_celsius(77))
print(fahrenheit_to_celsius(95))
print(fahrenheit_to_celsius(50))
````

#### Benefits:

- Code written once
- Used many times
- Easy to maintain

### Return Value
A return statement is used when a function needs to send a result back to the place where it was called.

````python
def get_greeting():
    return "Hello from a function"

message = get_greeting()
print(message)
````

### Step 1: Create the Function
```python
def get_greeting():
    return "Hello from a function"
    
Here we create a function named get_greeting.

Inside it we wrote:

return "Hello from a function"

Meaning:

Whenever this function runs, it will send back this text.
```
### Step 2: Call the Function
```python
message = get_greeting()
```

When Python runs this line:

1. Python calls the function
2. It goes inside the function
3. It sees return "Hello from a function"
4. That value comes back out of the function

So this line becomes:
```python
message = "Hello from a function"
```
Now the variable message stores that text.

### Step 3: Print the Value
```python
print(message)

Output:

Hello from a function
```

#### Important:
Without return, the function cannot give the result back.

````python
def multiply(a, b):
    return a * b

result = multiply(4, 5)
print(result)
````

````python
def add():
    return 5 + 3

result = add()
print(result)
````

### 8. Without return

````python
def add():
    print(5 + 3)

result = add()
print(result)
````

Because print() only shows the value, it does not send it back.

- print → shows output
- return → sends output back

### 9. The pass Statement

Functions cannot be empty.

If you want to define a function but write code later, use pass.

````python
def my_function():
    pass
````
