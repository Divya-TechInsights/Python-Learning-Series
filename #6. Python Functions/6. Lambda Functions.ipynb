## Python Lambda

A lambda function is a small anonymous function.

#### means:
Lambda = a small, one-line function without a name

A lambda function can take any number of arguments, but can only have one expression.

Note: Lambda is just a shortcut to write a function in one line.

### Normal Function

````python
def add(a):
    return a + 10

print(add(5))
````

### Now Convert to Lambda

````python
x = lambda a: a + 10
print(x(5))
````

Lambda does the same work as a function, but in one line and without a name.

### Break the Syntax
```python
lambda a: a + 10

Explain:

lambda → keyword
a → input
a + 10 → expression (result)
```
Say:

“Whatever is after : is automatically returned”

### Important Rule

Lambda can only have ONE expression
```python
Wrong:

lambda a:
    print(a)
    return a+1
```

### Multiple Arguments
Multiply argument a with argument b and return the result

````python
x = lambda a, b: a + b
print(x(5, 3))
````

Summarize argument a, b, and c and return the result

````python
x = lambda a, b, c : a + b + c
print(x(5, 6, 2))
````

### Real Purpose of Lambda
Lambda is useful when you want to create a quick (anonymous) function inside another function

````python
def myfunc(n):
  return lambda a: a * n
````

### Create a function that always doubles any number

````python
def myfunc(n):
  return lambda a: a * n

mydoubler = myfunc(2)

print(mydoubler(11))
````

### Create a function that triples the number

````python
def myfunc(n):
  return lambda a: a * n
  
mytripler = myfunc(3)
print(mytripler(10))   
````

#### Normally:

A function runs → finishes → variables disappear 

#### So:

“If myfunc is finished, how does n still exist?”

#### Simple Explanation 

“Python keeps n stored with the lambda function, so it can use it later.”

### Use the same function definition to make both functions, in the same program

````python
def myfunc(n):
  return lambda a : a * n

mydoubler = myfunc(2)
mytripler = myfunc(3)

print(mydoubler(11))
print(mytripler(11))
````

### Lambda with Built-in Functions
Lambda functions are commonly used with built-in functions like map(), filter(), and sorted().

### 1. map() – Transform each element
Applies a function to every item in an iterable (like list, tuple) and returns a new iterator.

````python
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

print(result)
````

````python
nums = [1, 2, 3]

result = list(map(lambda x: str(x), nums))
print(result)
````

### 2. filter() – Select specific elements
Filters elements based on a condition (True/False).

````python
numbers = [1, 2, 3, 4, 5]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
````

````python
words = ["cat", "elephant", "dog", "tiger"]

result = list(filter(lambda x: len(x) > 4, words))
print(result)
````

### 3. sorted() – Sort elements
Returns a new sorted list from an iterable.

````python
students = [("Aman", 25), ("Riya", 20), ("Karan", 23)]

result = sorted(students, key=lambda x: x[0])

print(result)
````

````python
words = ["grape", "kiwi", "banana", "apple"]

result = sorted(words, key=lambda x: len(x))
print(result)
````

| Function   | Purpose        | Lambda Role            |
| ---------- | -------------- | ---------------------- |
| `map()`    | Transform data | Defines transformation |
| `filter()` | Select data    | Defines condition      |
| `sorted()` | Arrange data   | Defines sorting key    |
