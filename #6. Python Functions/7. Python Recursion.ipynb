### Recursion
Recursion means a function calls itself to solve a smaller problem.

### How Recursion Actually Works

#### Recursion works in 2 phases:

- Going DOWN (Function calls itself)
- Coming UP (Returning answers)

### Example
- You ask friend A
- A asks B
- B asks C
- C gives answer
- Then answer travels back: C → B → A

### 2 Important parts:
#### 1. Base Case (STOP)

- “When should we stop?”

#### 2. Recursive Case (CALL)

- “Call the same function with smaller input”

### Factorial

````python
def factorial(n):
    if n == 1: #Base case
        return 1
    return n * factorial(n-1) #Recursive case

print(factorial(4))
````

```python
Step-by-Step Execution

-> GOING DOWN (calls keep stacking)
factorial(4)
→ 4 * factorial(3)

factorial(3)
→ 3 * factorial(2)

factorial(2)
→ 2 * factorial(1)

factorial(1)
→ 1   (Base case reached)


-> COMING BACK UP (returns happen)

Now values return:

factorial(2) = 2 * 1 = 2  
factorial(3) = 3 * 2 = 6  
factorial(4) = 4 * 6 = 24  

Final Answer: 24
```

### Sum of Numbers

````python
def sum(n):
    if n == 0:
        return 0
    return n + sum(n - 1)

print(sum(5))
````

### Fibonacci

````python
def fibonacci(n):
    if n <= 1:   # Base case
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6))
````

```python
Sequence:

0, 1, 1, 2, 3, 5, 8, ...

So:

fibonacci(6) = 8


Step 1: Expand Calls (Going DOWN)
fibonacci(6)
= fibonacci(5) + fibonacci(4)

Now expand further:

fibonacci(5)
= fibonacci(4) + fibonacci(3)
fibonacci(4)
= fibonacci(3) + fibonacci(2)
fibonacci(3)
= fibonacci(2) + fibonacci(1)
fibonacci(2)
= fibonacci(1) + fibonacci(0)

Base Cases
fibonacci(1) = 1
fibonacci(0) = 0

Step 2: Coming BACK UP (Solve values)

Now start solving from bottom:

fibonacci(2) = 1 + 0 = 1  
fibonacci(3) = 1 + 1 = 2  
fibonacci(4) = 2 + 1 = 3  
fibonacci(5) = 3 + 2 = 5  
fibonacci(6) = 5 + 3 = 8  

Final Answer
8
```

### Recursion with Lists

````python
def sum_list(numbers):
  if len(numbers) == 0:
    return 0
  else:
    return numbers[0] + sum_list(numbers[1:])

my_list = [1, 2, 3, 4, 5]
print(sum_list(my_list))
````

```python
Step 1: Going DOWN (list gets smaller)
sum_list([1,2,3,4,5])
= 1 + sum_list([2,3,4,5])

= 1 + (2 + sum_list([3,4,5]))

= 1 + (2 + (3 + sum_list([4,5])))

= 1 + (2 + (3 + (4 + sum_list([5]))))

= 1 + (2 + (3 + (4 + (5 + sum_list([])))))

Base Case
sum_list([]) = 0   

Empty list means nothing to add

Step 2: Coming UP (adding happens)
sum_list([5]) = 5 + 0 = 5  
sum_list([4,5]) = 4 + 5 = 9  
sum_list([3,4,5]) = 3 + 9 = 12  
sum_list([2,3,4,5]) = 2 + 12 = 14  
sum_list([1,2,3,4,5]) = 1 + 14 = 15  

Final Output
15
```

### Find the maximum value in a list

````python
def find_max(numbers):
  if len(numbers) == 1:
    return numbers[0]
  else:
    max_of_rest = find_max(numbers[1:])
    return numbers[0] if numbers[0] > max_of_rest else max_of_rest

my_list = [3, 7, 2, 9, 1]
print(find_max(my_list))
````

```python
Step 1: Going DOWN (list becomes smaller)
find_max([3,7,2,9,1])
→ compare 3 with max([7,2,9,1])

find_max([7,2,9,1])
→ compare 7 with max([2,9,1])

find_max([2,9,1])
→ compare 2 with max([9,1])

find_max([9,1])
→ compare 9 with max([1])

find_max([1])
→ 1    (Base case)

Step 2: Coming UP (comparisons happen)
find_max([9,1])
→ max(9, 1) = 9

find_max([2,9,1])
→ max(2, 9) = 9

find_max([7,2,9,1])
→ max(7, 9) = 9

find_max([3,7,2,9,1])
→ max(3, 9) = 9

Final Output
9
```

### What is Recursion Depth Limit?

Say this simply:

“Python allows only a limited number of recursive calls (around 1000).”

### Why does this limit exist?

Because of the call stack (memory)

- Every recursive call uses memory.
- Too many calls → memory overflow

So Python stops it to prevent crashes.

### What happens if limit is crossed?

You get this error:

RecursionError: maximum recursion depth exceeded

### Example (Understanding the Limit)

````python
import sys
print(sys.getrecursionlimit())
````

The recursion limit is not fixed. It depends on your Python environment

### Increasing the Limit

````python
import sys
sys.setrecursionlimit(5000)
print(sys.getrecursionlimit())
````
